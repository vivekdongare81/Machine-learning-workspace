```python
#Model representation

import numpy as np
import matplotlib.pyplot as plt

#plt.style.use('./deeplearning.mplstyle')

x_train = np.array([1,2]) #features
y_train = np.array([300,500]) #targets
print(f"Number of training example are : {x_train.shape[0]}")

#plot graph
plt.scatter(x_train, y_train, marker='x', c = 'r') #plot datapoints
plt.title("Housing proces")
plt.ylabel("Prices")
plt.xlabel("Size")
plt.show()
```

```python

#Model Function 
# Let initial w & b
w = 100
b = 100
# wx+b for each value
def compute_model_output(x, w, b):
    m = x.shape[0]
    f_wb = np.zeros(m) #np.zeros(m)  means create an array of zeros of size m
    for i in range(m):
        f_wb[i] = w*x[i] + b #predict
    return f_wb  

initial_f_wb = compute_model_output(x_train, w, b)
print(initial_f_wb) #this should be printing  
#      100*1 + 100 = 200
#      100*2 + 100 = 300

# Plot  model prediction
def plot_graph(x, y, predictions):
    plt.scatter(x, y, marker='x', c='r',label='Actual Values')
    plt.plot(x, predictions, c='b',label='Our Prediction')
    plt.title("Predictions")
    plt.ylabel('Target y')
    plt.xlabel('Input x')
    plt.legend()
    plt.show()

plot_graph(x_train, y_train, initial_f_wb)
```

```python

#Fix w and b manutally to see how our model performs
fixed_f_wb = compute_model_output(x_train, 200, 100)
print(fixed_f_wb) #this should be printing  

plot_graph(x_train, y_train, fixed_f_wb)
```

```python

# Cost Function

def compute_cost(x, y, w, b):
    m = x.shape[0]
    loss_sum = 0
    for i in range(m):
        predict = w*x[i] + b #y^
        loss = (predict - y[i]) **2 # (y^ - y)^2
        loss_sum = loss_sum + loss
    cost = loss_sum * (1/(2*m)) #1/2m SUM(y^ - y)^2
    return cost

cost = compute_cost(x_train, y_train, w, b)#original 100 100
print(cost)
cost = compute_cost(x_train, y_train, 200, 100)#Fixed 200 100
print(cost)
```

```python

# Gradient - return just 1 dj_dw for single feature + 1 jd_jb

def compute_gradient(x, y, w, b):

    m = x.shape[0]
    dj_dw = 0
    dj_db = 0
    for i in range(m):
        predict = w * x[i] + b
        loss = predict - y[i]
        dj_dw += loss * x[i]
        dj_db += loss
    dj_dw/=m
    dj_db/=m
    return dj_dw, dj_db

print("x_train:", x_train)
print("y_train:", y_train)
print("m =", x_train.shape[0])

alpha = 0.1   # learning rate

# Step 1
gradient_w, gradient_b = compute_gradient(x_train, y_train, w, b)
print(f"Step 1 -> Gradient: ({gradient_w}, {gradient_b})")

w = w - alpha * gradient_w
b = b - alpha * gradient_b
print(f"Updated w = {w}, b = {b}")
print(f"Cost = {compute_cost(x_train, y_train, w, b)}")
print()

# Step 2
gradient_w, gradient_b = compute_gradient(x_train, y_train, w, b)
print(f"Step 2 -> Gradient: ({gradient_w}, {gradient_b})")

w = w - alpha * gradient_w
b = b - alpha * gradient_b
print(f"Updated w = {w}, b = {b}")
print(f"Cost = {compute_cost(x_train, y_train, w, b)}")
print()

# Step 3
gradient_w, gradient_b = compute_gradient(x_train, y_train, w, b)
print(f"Step 3 -> Gradient: ({gradient_w}, {gradient_b})")

w = w - alpha * gradient_w
b = b - alpha * gradient_b
print(f"Updated w = {w}, b = {b}")
print(f"Cost = {compute_cost(x_train, y_train, w, b)}")
print()

# Step 4
gradient_w, gradient_b = compute_gradient(x_train, y_train, w, b)
print(f"Step 4 -> Gradient: ({gradient_w}, {gradient_b})")

w = w - alpha * gradient_w
b = b - alpha * gradient_b
print(f"Updated w = {w}, b = {b}")
print(f"Cost = {compute_cost(x_train, y_train, w, b)}")

```

```python
# Gradient Descent
import copy
import math 

def gradient_descent(x, y, w, b, alpha, epochs, compute_cost, compute_gradient):
    w = copy.deepcopy(w) #to avoid modifying orignal array
    b = copy.deepcopy(b) #to avoid modifying orignal array
    cost_history = []
    params_history = []

    for i in range(epochs):
        #Calc Gradients
        dj_dw, dj_db = compute_gradient(x, y, w, b)

        #Update Parameters
        w = w - alpha * dj_dw
        b = b - alpha * dj_db

        #Limit Resouces for saving data
        if i < 100000:
            cost = compute_cost(x, y, w, b)
            cost_history.append(cost)
            params_history.append([w,b])
        #Print status - {i:4} = 000i and {i:0.2e} = i.23e-12
        if i% math.ceil(epochs/10) == 0:# after each 10 percent of epochs
            print(f"Iteration : {i} Cost : {cost_history[-1]: 0.2e} dj_dw : {dj_dw: 0.2e} dj_db : {dj_db} w : {w: 0.2e} b: {b:0.2e}")

    return w, b, cost_history, params_history

w_init = 0
b_init = 0

epochs = 1000 #100000 stables at 0
alpha = 1.0e-2

initial_f_wb = compute_model_output(x_train, w_init, b_init)
plot_graph(x_train, y_train, initial_f_wb)  

w_final, b_final , cost_history, param_history = gradient_descent(x_train, y_train, w_init, b_init, alpha, epochs, compute_cost, compute_gradient)
print(f"Final (w,b) found by gradient descent : {w_final, b_final}")


#Fix w and b manutally to see how our model performs
final_f_wb = compute_model_output(x_train, w_final, b_final)
print(final_f_wb)

plot_graph(x_train, y_train, final_f_wb)  



```

```python
def plot_cost_history(cost_history):
    plt.figure(figsize=(6,4))

    plt.plot(cost_history)

    plt.title("Cost vs Iteration")
    plt.xlabel("Iteration")
    plt.ylabel("Cost J(w,b)")

    plt.grid(True)
    plt.show()

def plot_parameter_history(param_history):

    w_history = [p[0] for p in param_history]
    b_history = [p[1] for p in param_history]

    plt.figure(figsize=(6,4))

    plt.plot(w_history, label="w")
    plt.plot(b_history, label="b")

    plt.title("Parameter Convergence")
    plt.xlabel("Iteration")
    plt.ylabel("Value")

    plt.legend()
    plt.grid(True)
    plt.show()

def plot_gradient_descent_path(
        x,
        y,
        param_history,
        w_final,
        b_final,
        compute_cost,
        steps=None):

    # Extract w history
    w_history = [p[0] for p in param_history]

    # Cost at each visited w
    cost_history_w = []

    for w in w_history:
        cost_history_w.append(
            compute_cost(x, y, w, b_final)
        )

    # Smooth cost curve
    w_values = np.linspace(
        min(w_history) - 20,
        max(w_history) + 20,
        300
    )

    cost_values = []

    for w in w_values:
        cost_values.append(
            compute_cost(x, y, w, b_final)
        )

    if steps is None:
        steps = len(w_history)

    plt.figure(figsize=(8,6))

    # Cost curve
    plt.plot(
        w_values,
        cost_values,
        linewidth=2,
        color="blue",
        label="Cost Function J(w)"
    )

    # Gradient Descent arrows
    for i in range(min(steps-1, len(w_history)-1)):
        plt.annotate(
            "",
            xy=(w_history[i+1], cost_history_w[i+1]),
            xytext=(w_history[i], cost_history_w[i]),
            arrowprops=dict(
                arrowstyle="->",
                color="red",
                lw=1.5
            )
        )

    # Gradient Descent points
    plt.scatter(
        w_history[:steps],
        cost_history_w[:steps],
        color="red",
        s=35,
        label="Gradient Descent"
    )

    # Minimum
    plt.scatter(
        w_final,
        compute_cost(x, y, w_final, b_final),
        color="green",
        s=80,
        label="Minimum"
    )

    plt.title("Gradient Descent on Cost Function")
    plt.xlabel("w")
    plt.ylabel("J(w)")
    plt.grid(True)
    plt.legend()

    plt.show()


plot_gradient_descent_path(
    x_train,
    y_train,
    param_history,
    w_final,
    b_final,
    compute_cost,
    steps=None
)
plot_cost_history(cost_history)
plot_parameter_history(param_history)
```
