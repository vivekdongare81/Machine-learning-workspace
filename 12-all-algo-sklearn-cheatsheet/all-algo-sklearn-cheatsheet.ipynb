```python
"""
ML / SKLEARN USAGE CHEATSHEET
=============================

Purpose:
    One copyable Python file containing practical library usage for every algorithm
    we covered from Linear Regression -> Classical ML -> Dimensionality Reduction
    -> Anomaly Detection -> Recommenders -> Time Series / Forecasting.

Depth:
    This file focuses on LIBRARY / SKLEARN USAGE.
    It is intentionally NOT a from-scratch implementation file.

Install (as needed):
    pip install numpy pandas matplotlib scikit-learn scipy statsmodels
    pip install xgboost lightgbm catboost umap-learn

IMPORTANT:
    - XGBoost / LightGBM / CatBoost / UMAP are external libraries.
    - ARIMA/SARIMA/VAR are generally provided by statsmodels, not sklearn.
    - RNN/LSTM/GRU/Transformers/Chronos are NOT sklearn models and are therefore
      kept out of this sklearn-focused file.
"""

# ============================================================
# 0. COMMON IMPORTS + SAMPLE DATA
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, TimeSeriesSplit, GridSearchCV
from sklearn.preprocessing import StandardScaler, MinMaxScaler, PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    r2_score,
    silhouette_score,
)
from sklearn.datasets import (
    make_regression,
    make_classification,
    make_blobs,
    make_moons,
)

# Regression data
X_reg, y_reg = make_regression(
    n_samples=500,
    n_features=5,
    noise=15,
    random_state=42,
)

# Classification data
X_clf, y_clf = make_classification(
    n_samples=500,
    n_features=8,
    n_informative=5,
    n_redundant=1,
    n_classes=2,
    random_state=42,
)

# Clustering data
X_cluster, _ = make_blobs(
    n_samples=500,
    centers=4,
    cluster_std=1.2,
    random_state=42,
)

# Generic train/test split
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42,
)

X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X_clf,
    y_clf,
    test_size=0.2,
    random_state=42,
    stratify=y_clf,
)


# ============================================================
# 1. LINEAR REGRESSION
# ============================================================

from sklearn.linear_model import LinearRegression

linear_model = LinearRegression(
    fit_intercept=True,
    positive=False,
)

linear_model.fit(X_train_reg, y_train_reg)

y_pred = linear_model.predict(X_test_reg)

print("\nLINEAR REGRESSION")
print("MAE:", mean_absolute_error(y_test_reg, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test_reg, y_pred)))
print("R2:", r2_score(y_test_reg, y_pred))

# Important parameters:
#
# fit_intercept=True
#   Learns the intercept/bias term.
#
# positive=False
#   If True, forces all coefficients to be >= 0.
#
# Attributes worth knowing:
# coef_       -> learned feature weights
# intercept_  -> bias/intercept


# ============================================================
# 2. RIDGE REGRESSION
# ============================================================

from sklearn.linear_model import Ridge

ridge_model = Ridge(
    alpha=1.0,
    fit_intercept=True,
)

ridge_model.fit(X_train_reg, y_train_reg)

y_pred = ridge_model.predict(X_test_reg)

print("\nRIDGE")
print("RMSE:", np.sqrt(mean_squared_error(y_test_reg, y_pred)))
print("R2:", r2_score(y_test_reg, y_pred))

# Important parameters:
#
# alpha=1.0
#   L2 regularization strength.
#   Larger alpha -> stronger shrinkage.
#
# fit_intercept=True
#   Learns intercept.


# ============================================================
# 3. LASSO REGRESSION
# ============================================================

from sklearn.linear_model import Lasso

lasso_model = Lasso(
    alpha=0.1,
    max_iter=10000,
)

lasso_model.fit(X_train_reg, y_train_reg)

y_pred = lasso_model.predict(X_test_reg)

print("\nLASSO")
print("RMSE:", np.sqrt(mean_squared_error(y_test_reg, y_pred)))
print("R2:", r2_score(y_test_reg, y_pred))

# Important parameters:
#
# alpha=0.1
#   L1 regularization strength.
#   Larger alpha -> more coefficients pushed toward zero.
#
# max_iter=10000
#   Maximum optimization iterations.
#
# Main idea:
#   L1 regularization can perform feature selection because some coefficients
#   can become exactly zero.


# ============================================================
# 4. ELASTIC NET
# ============================================================

from sklearn.linear_model import ElasticNet

elastic_model = ElasticNet(
    alpha=0.1,
    l1_ratio=0.5,
    max_iter=10000,
)

elastic_model.fit(X_train_reg, y_train_reg)

y_pred = elastic_model.predict(X_test_reg)

print("\nELASTIC NET")
print("RMSE:", np.sqrt(mean_squared_error(y_test_reg, y_pred)))
print("R2:", r2_score(y_test_reg, y_pred))

# Important parameters:
#
# alpha
#   Overall regularization strength.
#
# l1_ratio
#   0   -> pure Ridge/L2
#   1   -> pure Lasso/L1
#   between -> mixture of L1 + L2


# ============================================================
# 5. POLYNOMIAL REGRESSION
# ============================================================

from sklearn.pipeline import make_pipeline

poly_model = make_pipeline(
    PolynomialFeatures(degree=2, include_bias=False),
    StandardScaler(),
    Ridge(alpha=1.0),
)

poly_model.fit(X_train_reg, y_train_reg)

y_pred = poly_model.predict(X_test_reg)

print("\nPOLYNOMIAL REGRESSION")
print("RMSE:", np.sqrt(mean_squared_error(y_test_reg, y_pred)))
print("R2:", r2_score(y_test_reg, y_pred))

# Important parameters:
#
# degree=2
#   Polynomial degree.
#
# include_bias=False
#   Avoids adding an extra constant column because Ridge already has an
#   intercept.
#
# Practical:
#   Higher degree can capture more nonlinear patterns but increases
#   overfitting risk.


# ============================================================
# 6. LOGISTIC REGRESSION
# ============================================================

from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        C=1.0,
        penalty="l2",
        solver="lbfgs",
        max_iter=1000,
    )),
])

logistic_model.fit(X_train_clf, y_train_clf)

y_pred = logistic_model.predict(X_test_clf)
y_prob = logistic_model.predict_proba(X_test_clf)[:, 1]

print("\nLOGISTIC REGRESSION")
print("Accuracy:", accuracy_score(y_test_clf, y_pred))
print("Precision:", precision_score(y_test_clf, y_pred))
print("Recall:", recall_score(y_test_clf, y_pred))
print("F1:", f1_score(y_test_clf, y_pred))

# Important parameters:
#
# C=1.0
#   Inverse regularization strength.
#   Smaller C -> stronger regularization.
#   Larger C -> weaker regularization.
#
# penalty="l2"
#   Regularization type.
#   Common choices: l1, l2, elasticnet depending on solver.
#
# solver="lbfgs"
#   Optimization algorithm.
#
# max_iter
#   Maximum optimization iterations.
#
# predict_proba()
#   Gives class probabilities.


# ============================================================
# 7. DECISION TREE
# ============================================================

from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor

dt_clf = DecisionTreeClassifier(
    criterion="gini",
    max_depth=5,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features=None,
    random_state=42,
)

dt_clf.fit(X_train_clf, y_train_clf)

y_pred = dt_clf.predict(X_test_clf)

print("\nDECISION TREE CLASSIFIER")
print("Accuracy:", accuracy_score(y_test_clf, y_pred))

# Important parameters:
#
# criterion="gini"
#   Split quality.
#   Classification: gini, entropy, log_loss.
#
# max_depth=5
#   Maximum tree depth.
#   Smaller -> simpler tree / more regularization.
#
# min_samples_split=2
#   Minimum samples required to split a node.
#
# min_samples_leaf=1
#   Minimum samples required in a leaf.
#
# max_features=None
#   Number of features considered at each split.
#
# random_state
#   Reproducibility.


dt_reg = DecisionTreeRegressor(
    criterion="squared_error",
    max_depth=5,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42,
)

dt_reg.fit(X_train_reg, y_train_reg)

y_pred = dt_reg.predict(X_test_reg)

print("\nDECISION TREE REGRESSOR")
print("RMSE:", np.sqrt(mean_squared_error(y_test_reg, y_pred)))


# ============================================================
# 8. RANDOM FOREST
# ============================================================

from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor

rf_clf = RandomForestClassifier(
    n_estimators=200,
    criterion="gini",
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    bootstrap=True,
    n_jobs=-1,
    random_state=42,
)

rf_clf.fit(X_train_clf, y_train_clf)

y_pred = rf_clf.predict(X_test_clf)

print("\nRANDOM FOREST")
print("Accuracy:", accuracy_score(y_test_clf, y_pred))

# Important parameters:
#
# n_estimators=200
#   Number of trees.
#
# max_depth=None
#   Maximum depth of each tree.
#
# max_features="sqrt"
#   Number of features considered per split.
#
# bootstrap=True
#   Each tree trains on a bootstrap sample.
#
# n_jobs=-1
#   Use all CPU cores.
#
# min_samples_leaf
#   Controls leaf size and overfitting.


rf_reg = RandomForestRegressor(
    n_estimators=200,
    max_depth=None,
    min_samples_leaf=1,
    max_features=1.0,
    n_jobs=-1,
    random_state=42,
)

rf_reg.fit(X_train_reg, y_train_reg)

y_pred = rf_reg.predict(X_test_reg)

print("RF Regression RMSE:", np.sqrt(mean_squared_error(y_test_reg, y_pred)))


# ============================================================
# 9. ADABOOST
# ============================================================

from sklearn.ensemble import AdaBoostClassifier, AdaBoostRegressor

ada_clf = AdaBoostClassifier(
    n_estimators=100,
    learning_rate=0.5,
    random_state=42,
)

ada_clf.fit(X_train_clf, y_train_clf)

y_pred = ada_clf.predict(X_test_clf)

print("\nADABOOST")
print("Accuracy:", accuracy_score(y_test_clf, y_pred))

# Important parameters:
#
# n_estimators
#   Number of weak learners.
#
# learning_rate
#   Contribution of each weak learner.
#   Lower value usually requires more estimators.
#
# random_state
#   Reproducibility.


# ============================================================
# 10. GRADIENT BOOSTING
# ============================================================

from sklearn.ensemble import GradientBoostingClassifier, GradientBoostingRegressor

gb_clf = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=3,
    min_samples_split=2,
    min_samples_leaf=1,
    subsample=1.0,
    random_state=42,
)

gb_clf.fit(X_train_clf, y_train_clf)

y_pred = gb_clf.predict(X_test_clf)

print("\nGRADIENT BOOSTING")
print("Accuracy:", accuracy_score(y_test_clf, y_pred))

# Important parameters:
#
# n_estimators
#   Number of boosting stages/trees.
#
# learning_rate
#   Shrinks contribution of each tree.
#
# max_depth
#   Complexity of individual trees.
#
# subsample
#   Fraction of samples used for each boosting stage.
#   < 1.0 introduces stochastic gradient boosting.


gb_reg = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=3,
    loss="squared_error",
    random_state=42,
)

gb_reg.fit(X_train_reg, y_train_reg)

y_pred = gb_reg.predict(X_test_reg)

print("GB Regression RMSE:", np.sqrt(mean_squared_error(y_test_reg, y_pred)))


# ============================================================
# 11. XGBOOST
# ============================================================

# pip install xgboost

try:
    from xgboost import XGBClassifier, XGBRegressor

    xgb_clf = XGBClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=6,
        min_child_weight=1,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.0,
        reg_lambda=1.0,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
    )

    xgb_clf.fit(X_train_clf, y_train_clf)

    y_pred = xgb_clf.predict(X_test_clf)

    print("\nXGBOOST")
    print("Accuracy:", accuracy_score(y_test_clf, y_pred))

    # Important parameters:
    #
    # n_estimators
    #   Number of boosting trees.
    #
    # learning_rate
    #   Step size for each tree.
    #
    # max_depth
    #   Tree depth.
    #
    # min_child_weight
    #   Minimum amount of weight needed in a child.
    #   Larger -> more conservative.
    #
    # subsample
    #   Fraction of rows used per tree.
    #
    # colsample_bytree
    #   Fraction of features used per tree.
    #
    # reg_alpha
    #   L1 regularization.
    #
    # reg_lambda
    #   L2 regularization.
    #
    # objective
    #   Learning objective, e.g. binary:logistic for binary classification.

except ImportError:
    print("\nXGBoost not installed -> pip install xgboost")


# ============================================================
# 12. LIGHTGBM
# ============================================================

# pip install lightgbm

try:
    from lightgbm import LGBMClassifier, LGBMRegressor

    lgbm_clf = LGBMClassifier(
        n_estimators=200,
        learning_rate=0.05,
        num_leaves=31,
        max_depth=-1,
        min_child_samples=20,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.0,
        reg_lambda=0.0,
        random_state=42,
        verbosity=-1,
    )

    lgbm_clf.fit(X_train_clf, y_train_clf)

    y_pred = lgbm_clf.predict(X_test_clf)

    print("\nLIGHTGBM")
    print("Accuracy:", accuracy_score(y_test_clf, y_pred))

    # Important parameters:
    #
    # n_estimators
    #   Number of boosting iterations.
    #
    # learning_rate
    #   Contribution of each tree.
    #
    # num_leaves
    #   Main complexity control in LightGBM.
    #   More leaves -> more complex model.
    #
    # max_depth
    #   Maximum tree depth. -1 means no explicit limit.
    #
    # min_child_samples
    #   Minimum samples in a leaf.
    #
    # subsample
    #   Row sampling.
    #
    # colsample_bytree
    #   Feature sampling.
    #
    # reg_alpha / reg_lambda
    #   L1 / L2 regularization.

except ImportError:
    print("\nLightGBM not installed -> pip install lightgbm")


# ============================================================
# 13. CATBOOST
# ============================================================

# pip install catboost

try:
    from catboost import CatBoostClassifier, CatBoostRegressor

    cat_clf = CatBoostClassifier(
        iterations=200,
        learning_rate=0.05,
        depth=6,
        l2_leaf_reg=3.0,
        loss_function="Logloss",
        verbose=False,
        random_seed=42,
    )

    cat_clf.fit(X_train_clf, y_train_clf)

    y_pred = cat_clf.predict(X_test_clf).ravel()

    print("\nCATBOOST")
    print("Accuracy:", accuracy_score(y_test_clf, y_pred))

    # Important parameters:
    #
    # iterations
    #   Number of boosting iterations.
    #
    # learning_rate
    #   Step size.
    #
    # depth
    #   Tree depth.
    #
    # l2_leaf_reg
    #   L2 regularization.
    #
    # loss_function
    #   Training objective.
    #
    # verbose
    #   Controls training output.
    #
    # random_seed
    #   Reproducibility.
    #
    # cat_features
    #   Important CatBoost feature:
    #   pass categorical column indices here when using categorical data.

except ImportError:
    print("\nCatBoost not installed -> pip install catboost")


# ============================================================
# 14. SUPPORT VECTOR MACHINE (SVM)
# ============================================================

from sklearn.svm import SVC, SVR

svm_clf = Pipeline([
    ("scaler", StandardScaler()),
    ("model", SVC(
        C=1.0,
        kernel="rbf",
        gamma="scale",
        degree=3,
        probability=True,
    )),
])

svm_clf.fit(X_train_clf, y_train_clf)

y_pred = svm_clf.predict(X_test_clf)

print("\nSVM")
print("Accuracy:", accuracy_score(y_test_clf, y_pred))

# Important parameters:
#
# C=1.0
#   Controls margin violation.
#   Large C -> tries harder to classify training points correctly.
#   Small C -> stronger regularization / wider margin.
#
# kernel="rbf"
#   Kernel function.
#   Common: linear, poly, rbf, sigmoid.
#
# gamma="scale"
#   Controls influence/radius of each point for RBF/poly/sigmoid.
#
# degree=3
#   Degree for polynomial kernel.
#
# probability=True
#   Enables probability estimates.
#   More expensive training.


svm_reg = Pipeline([
    ("scaler", StandardScaler()),
    ("model", SVR(
        C=10.0,
        kernel="rbf",
        gamma="scale",
        epsilon=0.1,
    )),
])

svm_reg.fit(X_train_reg, y_train_reg)

y_pred = svm_reg.predict(X_test_reg)

print("SVR RMSE:", np.sqrt(mean_squared_error(y_test_reg, y_pred)))

# SVR-specific:
#
# epsilon
#   Defines the epsilon-insensitive tube around the regression function.


# ============================================================
# 15. K-NEAREST NEIGHBORS
# ============================================================

from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor

knn_clf = Pipeline([
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier(
        n_neighbors=5,
        weights="distance",
        metric="minkowski",
        p=2,
    )),
])

knn_clf.fit(X_train_clf, y_train_clf)

y_pred = knn_clf.predict(X_test_clf)

print("\nKNN")
print("Accuracy:", accuracy_score(y_test_clf, y_pred))

# Important parameters:
#
# n_neighbors=5
#   Number of nearest points used.
#
# weights="distance"
#   Closer points get more influence.
#   "uniform" gives equal weight.
#
# metric="minkowski"
#   Distance metric family.
#
# p=2
#   p=2 -> Euclidean distance.
#   p=1 -> Manhattan distance.
#
# Scaling is IMPORTANT because KNN depends on distance.


# ============================================================
# 16. NAIVE BAYES
# ============================================================

from sklearn.naive_bayes import GaussianNB

nb_model = GaussianNB(
    var_smoothing=1e-9,
)

nb_model.fit(X_train_clf, y_train_clf)

y_pred = nb_model.predict(X_test_clf)

print("\nGAUSSIAN NAIVE BAYES")
print("Accuracy:", accuracy_score(y_test_clf, y_pred))

# Important parameter:
#
# var_smoothing=1e-9
#   Adds a small value to variances for numerical stability.
#
# Other variants:
#   GaussianNB -> continuous approximately Gaussian features
#   MultinomialNB -> count features
#   BernoulliNB -> binary features
#   ComplementNB -> often useful for imbalanced text classification


# ============================================================
# 17. LDA CLASSIFICATION
# ============================================================

from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

lda_clf = LinearDiscriminantAnalysis(
    solver="svd",
)

lda_clf.fit(X_train_clf, y_train_clf)

y_pred = lda_clf.predict(X_test_clf)

print("\nLDA CLASSIFIER")
print("Accuracy:", accuracy_score(y_test_clf, y_pred))

# Important parameters:
#
# solver="svd"
#   Common solver; does not require explicit shrinkage.
#
# Other solver:
#   "lsqr"
#   "eigen"
#
# shrinkage
#   Can be used with lsqr/eigen.
#   Helps when covariance estimation is unstable.


# ============================================================
# 18. QDA CLASSIFICATION
# ============================================================

from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis

qda_clf = QuadraticDiscriminantAnalysis(
    reg_param=0.0,
)

qda_clf.fit(X_train_clf, y_train_clf)

y_pred = qda_clf.predict(X_test_clf)

print("\nQDA")
print("Accuracy:", accuracy_score(y_test_clf, y_pred))

# Important parameter:
#
# reg_param=0.0
#   Regularizes class-specific covariance matrices.
#   Larger -> more regularization.


# ============================================================
# 19. GAUSSIAN MIXTURE MODEL (GMM)
# ============================================================

from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(
    n_components=4,
    covariance_type="full",
    max_iter=100,
    n_init=5,
    random_state=42,
)

gmm.fit(X_cluster)

cluster_labels = gmm.predict(X_cluster)
cluster_probabilities = gmm.predict_proba(X_cluster)

print("\nGMM")
print("Cluster labels:", cluster_labels[:10])

# Important parameters:
#
# n_components=4
#   Number of Gaussian components.
#
# covariance_type="full"
#   Each component gets its own full covariance matrix.
#
# Options:
#   full, tied, diag, spherical
#
# max_iter
#   Maximum EM iterations.
#
# n_init
#   Number of initializations.
#
# predict_proba()
#   Soft cluster assignment probabilities.


# ============================================================
# 20. K-MEANS
# ============================================================

from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=4,
    init="k-means++",
    n_init=10,
    max_iter=300,
    tol=1e-4,
    random_state=42,
)

kmeans.fit(X_cluster)

labels = kmeans.labels_
centers = kmeans.cluster_centers_

print("\nK-MEANS")
print("Silhouette:", silhouette_score(X_cluster, labels))

# Important parameters:
#
# n_clusters
#   Number of clusters K.
#
# init="k-means++"
#   Smart initialization of centroids.
#
# n_init=10
#   Number of different initializations.
#
# max_iter=300
#   Maximum iterations per initialization.
#
# tol
#   Convergence tolerance.
#
# cluster_centers_
#   Final centroids.


# ============================================================
# 21. HIERARCHICAL / AGGLOMERATIVE CLUSTERING
# ============================================================

from sklearn.cluster import AgglomerativeClustering

hierarchical = AgglomerativeClustering(
    n_clusters=4,
    metric="euclidean",
    linkage="ward",
)

hier_labels = hierarchical.fit_predict(X_cluster)

print("\nHIERARCHICAL CLUSTERING")
print("Silhouette:", silhouette_score(X_cluster, hier_labels))

# Important parameters:
#
# n_clusters
#   Number of final clusters.
#
# metric="euclidean"
#   Distance metric.
#
# linkage="ward"
#   How clusters are merged.
#
# Common linkage:
#   ward, complete, average, single
#
# Note:
#   Ward works with Euclidean distance.


# ============================================================
# 22. DBSCAN
# ============================================================

from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=0.5,
    min_samples=5,
    metric="euclidean",
)

db_labels = dbscan.fit_predict(X_cluster)

print("\nDBSCAN")
print("Labels:", np.unique(db_labels))

# Important parameters:
#
# eps=0.5
#   Radius around each point.
#
# min_samples=5
#   Minimum points needed to form a dense region.
#
# label -1
#   Usually represents noise/outliers.
#
# DBSCAN does not require choosing K.


# ============================================================
# 23. PCA
# ============================================================

from sklearn.decomposition import PCA

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_reg)

pca = PCA(
    n_components=2,
    whiten=False,
    svd_solver="auto",
    random_state=42,
)

X_pca = pca.fit_transform(X_scaled)

print("\nPCA")
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total explained variance:", pca.explained_variance_ratio_.sum())

# Important parameters:
#
# n_components=2
#   Number of dimensions to keep.
#
#   Can also use:
#   n_components=0.95
#   -> keep enough components to explain 95% variance.
#
# whiten=False
#   If True, scales transformed components to unit variance.
#
# svd_solver="auto"
#   SVD implementation selection.
#
# IMPORTANT:
#   Usually scale features before PCA when feature units differ.


# ============================================================
# 24. LDA DIMENSIONALITY REDUCTION
# ============================================================

# LDA can also be used as supervised dimensionality reduction.
# It uses class labels while PCA does not.

X_train_scaled = StandardScaler().fit_transform(X_train_clf)

lda_dr = LinearDiscriminantAnalysis(
    n_components=1,
    solver="svd",
)

X_lda = lda_dr.fit_transform(X_train_scaled, y_train_clf)

print("\nLDA DIMENSIONALITY REDUCTION")
print("Shape:", X_lda.shape)

# Important parameters:
#
# n_components
#   Maximum is min(number_of_features, number_of_classes - 1).
#
# Key difference:
#   PCA -> maximize variance.
#   LDA -> maximize class separation.


# ============================================================
# 25. t-SNE
# ============================================================

from sklearn.manifold import TSNE

X_tsne = TSNE(
    n_components=2,
    perplexity=30,
    learning_rate="auto",
    max_iter=1000,
    init="pca",
    random_state=42,
).fit_transform(X_scaled)

print("\nt-SNE")
print("Shape:", X_tsne.shape)

# Important parameters:
#
# n_components=2
#   Usually 2D visualization.
#
# perplexity=30
#   Roughly controls local neighborhood size.
#
# learning_rate
#   Optimization step size.
#
# max_iter
#   Number of optimization iterations.
#
# init="pca"
#   Initial embedding.
#
# IMPORTANT:
#   t-SNE is mainly for visualization, not a general-purpose feature
#   transformation pipeline.


# ============================================================
# 26. UMAP
# ============================================================

# pip install umap-learn

try:
    import umap

    umap_model = umap.UMAP(
        n_components=2,
        n_neighbors=15,
        min_dist=0.1,
        metric="euclidean",
        random_state=42,
    )

    X_umap = umap_model.fit_transform(X_scaled)

    print("\nUMAP")
    print("Shape:", X_umap.shape)

    # Important parameters:
    #
    # n_components=2
    #   Output dimensions.
    #
    # n_neighbors=15
    #   Local neighborhood size.
    #   Smaller -> more local structure.
    #   Larger -> more global structure.
    #
    # min_dist=0.1
    #   Minimum distance between embedded points.
    #   Smaller -> tighter clusters.
    #
    # metric="euclidean"
    #   Input distance metric.

except ImportError:
    print("\nUMAP not installed -> pip install umap-learn")


# ============================================================
# 27. ISOLATION FOREST
# ============================================================

from sklearn.ensemble import IsolationForest

iso = IsolationForest(
    n_estimators=200,
    contamination="auto",
    max_samples="auto",
    max_features=1.0,
    random_state=42,
)

iso.fit(X_reg)

iso_labels = iso.predict(X_reg)

print("\nISOLATION FOREST")
print("Number of detected outliers:", np.sum(iso_labels == -1))

# Important parameters:
#
# n_estimators
#   Number of isolation trees.
#
# contamination
#   Expected proportion of outliers.
#
# max_samples
#   Number of samples used to build each tree.
#
# max_features
#   Number/fraction of features used.
#
# predict():
#   +1 -> normal
#   -1 -> anomaly


# ============================================================
# 28. LOCAL OUTLIER FACTOR (LOF)
# ============================================================

from sklearn.neighbors import LocalOutlierFactor

lof = LocalOutlierFactor(
    n_neighbors=20,
    contamination=0.05,
    metric="minkowski",
)

lof_labels = lof.fit_predict(X_reg)

print("\nLOF")
print("Number of detected outliers:", np.sum(lof_labels == -1))

# Important parameters:
#
# n_neighbors
#   Neighborhood size used to estimate local density.
#
# contamination
#   Expected outlier fraction.
#
# metric
#   Distance metric.
#
# LOF compares local density of a point to density of its neighbors.


# ============================================================
# 29. ONE-CLASS SVM
# ============================================================

from sklearn.svm import OneClassSVM

one_class_svm = Pipeline([
    ("scaler", StandardScaler()),
    ("model", OneClassSVM(
        kernel="rbf",
        gamma="scale",
        nu=0.05,
    )),
])

one_class_svm.fit(X_reg)

ocs_labels = one_class_svm.predict(X_reg)

print("\nONE-CLASS SVM")
print("Number of detected outliers:", np.sum(ocs_labels == -1))

# Important parameters:
#
# kernel
#   Boundary function.
#
# gamma
#   Influence of each point for RBF kernel.
#
# nu
#   Upper bound on training errors and lower bound on support vectors.
#   Often treated as an approximate outlier fraction control.


# ============================================================
# 30. RECOMMENDER SYSTEM - USER-BASED COLLABORATIVE FILTERING
# ============================================================

from sklearn.metrics.pairwise import cosine_similarity

# Rows = users
# Columns = items
# Values = ratings / interactions

ratings = np.array([
    [5, 4, 0, 0, 2],
    [5, 5, 0, 1, 0],
    [0, 0, 5, 4, 5],
    [1, 0, 5, 4, 4],
    [5, 4, 1, 0, 0],
], dtype=float)

user_similarity = cosine_similarity(ratings)

print("\nUSER SIMILARITY")
print(user_similarity)

# Core library usage:
#
# cosine_similarity(A, B)
#   Computes cosine similarity between vectors.
#
# Higher value -> users have more similar interaction patterns.


# ============================================================
# 31. ITEM-BASED COLLABORATIVE FILTERING
# ============================================================

item_similarity = cosine_similarity(ratings.T)

print("\nITEM SIMILARITY")
print(item_similarity)

# ratings.T
#   Transpose so each item becomes a row.
#
# Then cosine similarity compares items instead of users.


# ============================================================
# 32. CONTENT-BASED RECOMMENDATION WITH TF-IDF
# ============================================================

from sklearn.feature_extraction.text import TfidfVectorizer

items = pd.DataFrame({
    "title": [
        "Java Spring Boot Microservices",
        "Python Machine Learning",
        "Deep Learning with Transformers",
        "Kubernetes DevOps",
        "Machine Learning Systems",
    ],
    "description": [
        "java spring boot backend microservices kafka",
        "python machine learning regression classification",
        "deep learning neural networks transformers attention",
        "kubernetes docker devops deployment cloud",
        "machine learning mlops production deployment models",
    ],
})

tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=1,
    max_df=1.0,
)

item_vectors = tfidf.fit_transform(items["description"])

content_similarity = cosine_similarity(item_vectors)

print("\nCONTENT-BASED RECOMMENDATION")
print(content_similarity)

# Important TF-IDF parameters:
#
# lowercase=True
#   Convert text to lowercase.
#
# stop_words="english"
#   Remove common English words.
#
# ngram_range=(1, 2)
#   Use unigrams + bigrams.
#
# min_df
#   Ignore terms appearing in too few documents.
#
# max_df
#   Ignore terms appearing in too many documents.
#
# Then:
#   cosine_similarity(item_vectors)
#   gives content similarity.


# ============================================================
# 33. RECOMMENDER - MATRIX FACTORIZATION CONCEPT
# ============================================================

# sklearn does not provide the same simple "production recommender" API
# as tree/classification models.
#
# A common library building block is TruncatedSVD for latent factors.

from sklearn.decomposition import TruncatedSVD

svd = TruncatedSVD(
    n_components=2,
    n_iter=10,
    random_state=42,
)

user_latent = svd.fit_transform(ratings)
item_latent = svd.components_.T

reconstructed_ratings = user_latent @ item_latent.T

print("\nMATRIX FACTORIZATION / LATENT FACTORS")
print(reconstructed_ratings)

# Important parameters:
#
# n_components
#   Number of latent dimensions.
#
# n_iter
#   Number of power iterations.
#
# random_state
#   Reproducibility.
#
# Concept:
#   users -> latent vectors
#   items -> latent vectors
#   dot product -> predicted preference


# ============================================================
# 34. RECOMMENDER - PRECISION@K / RECALL@K
# ============================================================

def precision_at_k(recommended, relevant, k):
    recommended = recommended[:k]
    relevant = set(relevant)

    if k == 0:
        return 0.0

    hits = sum(item in relevant for item in recommended)
    return hits / k


def recall_at_k(recommended, relevant, k):
    recommended = recommended[:k]
    relevant = set(relevant)

    if len(relevant) == 0:
        return 0.0

    hits = sum(item in relevant for item in recommended)
    return hits / len(relevant)


recommended_items = [3, 7, 2, 9, 5]
relevant_items = [2, 3, 5]

print("\nRECOMMENDER METRICS")
print("Precision@3:", precision_at_k(recommended_items, relevant_items, 3))
print("Recall@3:", recall_at_k(recommended_items, relevant_items, 3))

# Precision@K:
#   Of the top K recommended items, how many were relevant?
#
# Recall@K:
#   Of all relevant items, how many did we retrieve in top K?


# ============================================================
# 35. TIME SERIES - BASIC SETUP
# ============================================================

from datetime import datetime

np.random.seed(42)

dates = pd.date_range(
    start="2024-01-01",
    periods=300,
    freq="D",
)

trend = np.linspace(10, 30, len(dates))
seasonality = 5 * np.sin(2 * np.pi * np.arange(len(dates)) / 7)
noise = np.random.normal(0, 1.5, len(dates))

series = pd.Series(
    trend + seasonality + noise,
    index=dates,
    name="y",
)

print("\nTIME SERIES")
print(series.head())

series.plot(figsize=(12, 4), title="Time Series")
plt.tight_layout()
plt.show()


# ============================================================
# 36. LAG FEATURES
# ============================================================

ts_df = pd.DataFrame({"y": series})

ts_df["lag_1"] = ts_df["y"].shift(1)
ts_df["lag_7"] = ts_df["y"].shift(7)
ts_df["lag_14"] = ts_df["y"].shift(14)

ts_df["rolling_mean_7"] = ts_df["y"].shift(1).rolling(7).mean()
ts_df["rolling_std_7"] = ts_df["y"].shift(1).rolling(7).std()

ts_df["day_of_week"] = ts_df.index.dayofweek
ts_df["month"] = ts_df.index.month

ts_df = ts_df.dropna()

print("\nTIME-SERIES FEATURES")
print(ts_df.head())

# Important:
#
# shift(1)
#   Uses previous observation.
#
# shift(7)
#   Uses value from 7 time steps ago.
#
# rolling(7)
#   7-step rolling window.
#
# IMPORTANT:
#   Shift before rolling when creating prediction features.
#   Otherwise current/future information can leak into the model.


# ============================================================
# 37. TIME-SERIES TRAIN / TEST SPLIT
# ============================================================

split_index = int(len(ts_df) * 0.8)

train_ts = ts_df.iloc[:split_index]
test_ts = ts_df.iloc[split_index:]

feature_cols = [
    "lag_1",
    "lag_7",
    "lag_14",
    "rolling_mean_7",
    "rolling_std_7",
    "day_of_week",
    "month",
]

X_train_ts = train_ts[feature_cols]
y_train_ts = train_ts["y"]

X_test_ts = test_ts[feature_cols]
y_test_ts = test_ts["y"]

print("\nTIME SERIES SPLIT")
print("Train:", train_ts.index.min(), "->", train_ts.index.max())
print("Test :", test_ts.index.min(), "->", test_ts.index.max())


# ============================================================
# 38. NAIVE FORECAST
# ============================================================

naive_prediction = np.repeat(
    y_train_ts.iloc[-1],
    len(y_test_ts),
)

print("\nNAIVE FORECAST")
print("MAE:", mean_absolute_error(y_test_ts, naive_prediction))
print("RMSE:", np.sqrt(mean_squared_error(y_test_ts, naive_prediction)))

# Naive forecast:
#   future value = last observed value.


# ============================================================
# 39. SEASONAL NAIVE FORECAST
# ============================================================

season_length = 7

seasonal_values = list(y_train_ts.iloc[-season_length:])

seasonal_prediction = np.array([
    seasonal_values[i % season_length]
    for i in range(len(y_test_ts))
])

print("\nSEASONAL NAIVE")
print("MAE:", mean_absolute_error(y_test_ts, seasonal_prediction))
print("RMSE:", np.sqrt(mean_squared_error(y_test_ts, seasonal_prediction)))

# Seasonal naive:
#   future value repeats the value from the same seasonal position.


# ============================================================
# 40. EXPONENTIAL SMOOTHING / HOLT-WINTERS
# ============================================================

from statsmodels.tsa.holtwinters import ExponentialSmoothing

holt_winters = ExponentialSmoothing(
    y_train_ts,
    trend="add",
    seasonal="add",
    seasonal_periods=7,
)

holt_winters_fit = holt_winters.fit()

hw_forecast = holt_winters_fit.forecast(
    len(y_test_ts)
)

print("\nHOLT-WINTERS")
print("MAE:", mean_absolute_error(y_test_ts, hw_forecast))
print("RMSE:", np.sqrt(mean_squared_error(y_test_ts, hw_forecast)))

# Important parameters:
#
# trend="add"
#   Additive trend.
#
# trend="mul"
#   Multiplicative trend.
#
# seasonal="add"
#   Additive seasonality.
#
# seasonal="mul"
#   Multiplicative seasonality.
#
# seasonal_periods=7
#   Number of observations in one seasonal cycle.


# ============================================================
# 41. ACF
# ============================================================

from statsmodels.tsa.stattools import acf, pacf

acf_values = acf(
    series,
    nlags=24,
)

print("\nACF FIRST 13 VALUES")
for lag, value in enumerate(acf_values[:13]):
    print(f"Lag {lag}: {value:.4f}")

# Important:
#
# nlags
#   Number of lag values to calculate.
#
# ACF:
#   correlation between y_t and y_(t-k).


# ============================================================
# 42. PACF
# ============================================================

pacf_values = pacf(
    series,
    nlags=24,
    method="ywm",
)

print("\nPACF FIRST 13 VALUES")
for lag, value in enumerate(pacf_values[:13]):
    print(f"Lag {lag}: {value:.4f}")

# Important:
#
# PACF measures correlation at a lag after removing effects
# of intermediate lags.


# ============================================================
# 43. STATIONARITY - ADF TEST
# ============================================================

from statsmodels.tsa.stattools import adfuller

adf_result = adfuller(series.dropna())

print("\nADF TEST")
print("ADF statistic:", adf_result[0])
print("p-value:", adf_result[1])

# Interpretation:
#
# H0: series has a unit root / is non-stationary.
#
# Small p-value:
#   evidence against H0 -> often treated as stationary.
#
# Large p-value:
#   insufficient evidence to reject H0.


# ============================================================
# 44. DIFFERENCING
# ============================================================

differenced = series.diff().dropna()

print("\nDIFFERENCED SERIES")
print(differenced.head())

# diff():
#   y'_t = y_t - y_(t-1)
#
# Used to remove trend / make a series more stationary.


# ============================================================
# 45. AR MODEL
# ============================================================

from statsmodels.tsa.ar_model import AutoReg

ar_model = AutoReg(
    y_train_ts,
    lags=7,
    trend="c",
)

ar_fit = ar_model.fit()

ar_forecast = ar_fit.predict(
    start=len(y_train_ts),
    end=len(y_train_ts) + len(y_test_ts) - 1,
)

print("\nAR MODEL")
print("MAE:", mean_absolute_error(y_test_ts, ar_forecast))
print("RMSE:", np.sqrt(mean_squared_error(y_test_ts, ar_forecast)))

# Important parameters:
#
# lags=7
#   Number of previous observations used.
#
# trend="c"
#   Constant/intercept.
#
# AR idea:
#   y_t depends on previous y values.


# ============================================================
# 46. MA / ARMA CONCEPT
# ============================================================

# statsmodels does not require a separate simple "MA" class in the same
# way as sklearn estimators. ARMA can be represented as ARIMA(p,0,q).

from statsmodels.tsa.arima.model import ARIMA

arma_model = ARIMA(
    y_train_ts,
    order=(2, 0, 2),
)

arma_fit = arma_model.fit()

arma_forecast = arma_fit.forecast(
    steps=len(y_test_ts)
)

print("\nARMA")
print("MAE:", mean_absolute_error(y_test_ts, arma_forecast))
print("RMSE:", np.sqrt(mean_squared_error(y_test_ts, arma_forecast)))

# order=(p,d,q)
#
# p -> AR order
# d -> differencing
# q -> MA order
#
# ARMA is ARIMA with d=0.


# ============================================================
# 47. ARIMA
# ============================================================

arima_model = ARIMA(
    y_train_ts,
    order=(2, 1, 2),
)

arima_fit = arima_model.fit()

arima_forecast = arima_fit.forecast(
    steps=len(y_test_ts)
)

print("\nARIMA")
print("MAE:", mean_absolute_error(y_test_ts, arima_forecast))
print("RMSE:", np.sqrt(mean_squared_error(y_test_ts, arima_forecast)))

# Important parameters:
#
# order=(p,d,q)
#
# p -> number of AR terms
# d -> number of differences
# q -> number of MA terms
#
# Example:
#   ARIMA(2,1,2)
#   = 2 AR + 1 differencing + 2 MA.


# ============================================================
# 48. SARIMA / SARIMAX
# ============================================================

from statsmodels.tsa.statespace.sarimax import SARIMAX

sarima_model = SARIMAX(
    y_train_ts,
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 7),
)

sarima_fit = sarima_model.fit(disp=False)

sarima_forecast = sarima_fit.forecast(
    steps=len(y_test_ts)
)

print("\nSARIMA")
print("MAE:", mean_absolute_error(y_test_ts, sarima_forecast))
print("RMSE:", np.sqrt(mean_squared_error(y_test_ts, sarima_forecast)))

# Important parameters:
#
# order=(p,d,q)
#   Non-seasonal ARIMA part.
#
# seasonal_order=(P,D,Q,s)
#
# P -> seasonal AR order
# D -> seasonal differencing
# Q -> seasonal MA order
# s -> seasonal period
#
# Example:
#   seasonal_order=(1,1,1,7)
#   means weekly seasonality for daily data.


# ============================================================
# 49. SARIMAX WITH EXOGENOUS VARIABLES
# ============================================================

# Example exogenous feature:
# day_of_week

exog_train = train_ts[["day_of_week"]]
exog_test = test_ts[["day_of_week"]]

sarimax_model = SARIMAX(
    y_train_ts,
    exog=exog_train,
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 7),
)

sarimax_fit = sarimax_model.fit(disp=False)

sarimax_forecast = sarimax_fit.forecast(
    steps=len(y_test_ts),
    exog=exog_test,
)

print("\nSARIMAX")
print("MAE:", mean_absolute_error(y_test_ts, sarimax_forecast))
print("RMSE:", np.sqrt(mean_squared_error(y_test_ts, sarimax_forecast)))

# exog:
#   External variables that help explain the target.
#
# Important:
#   Future exogenous values must be known or forecast separately.


# ============================================================
# 50. VAR - VECTOR AUTOREGRESSION
# ============================================================

from statsmodels.tsa.api import VAR

multi_ts = pd.DataFrame({
    "y1": series,
    "y2": series.shift(1).bfill() + np.random.normal(0, 1, len(series)),
}, index=series.index)

var_train = multi_ts.iloc[:-30]
var_test = multi_ts.iloc[-30:]

var_model = VAR(
    var_train
)

var_fit = var_model.fit(
    maxlags=7,
    ic="aic",
)

var_forecast = var_fit.forecast(
    var_train.values[-var_fit.k_ar:],
    steps=len(var_test),
)

var_forecast_df = pd.DataFrame(
    var_forecast,
    index=var_test.index,
    columns=var_test.columns,
)

print("\nVAR")
print(var_forecast_df.head())

# Important:
#
# maxlags=7
#   Maximum lag order considered.
#
# ic="aic"
#   Information criterion used to select lag order.
#
# VAR:
#   Multiple time series predict each other using their past values.


# ============================================================
# 51. VARX CONCEPT
# ============================================================

# VARX = VAR + external/exogenous variables.
#
# statsmodels' VAR class handles multivariate endogenous series.
# External variables can be incorporated with a state-space / regression
# formulation depending on the exact production design.
#
# Conceptually:
#
#   y1_t = past(y1, y2) + external_x_t
#   y2_t = past(y1, y2) + external_x_t
#
# For production implementation, choose the statsmodels model/API based
# on the exact exogenous-variable structure.


# ============================================================
# 52. ML FORECASTING WITH RANDOM FOREST
# ============================================================

from sklearn.ensemble import RandomForestRegressor

rf_forecast_model = RandomForestRegressor(
    n_estimators=300,
    max_depth=10,
    min_samples_leaf=2,
    max_features=1.0,
    random_state=42,
    n_jobs=-1,
)

rf_forecast_model.fit(
    X_train_ts,
    y_train_ts,
)

rf_forecast = rf_forecast_model.predict(
    X_test_ts,
)

print("\nML FORECASTING - RANDOM FOREST")
print("MAE:", mean_absolute_error(y_test_ts, rf_forecast))
print("RMSE:", np.sqrt(mean_squared_error(y_test_ts, rf_forecast)))

# Key idea:
#
# Convert time series into supervised ML:
#
#   lag_1
#   lag_7
#   rolling mean
#   calendar features
#       ->
#   target y_t
#
# Important:
#   Never randomly shuffle a time-series split.


# ============================================================
# 53. ML FORECASTING WITH GRADIENT BOOSTING
# ============================================================

gb_forecast_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    min_samples_leaf=3,
    random_state=42,
)

gb_forecast_model.fit(
    X_train_ts,
    y_train_ts,
)

gb_forecast = gb_forecast_model.predict(
    X_test_ts,
)

print("\nML FORECASTING - GRADIENT BOOSTING")
print("MAE:", mean_absolute_error(y_test_ts, gb_forecast))
print("RMSE:", np.sqrt(mean_squared_error(y_test_ts, gb_forecast)))


# ============================================================
# 54. TIME-SERIES CROSS VALIDATION
# ============================================================

tscv = TimeSeriesSplit(
    n_splits=5,
)

print("\nTIME SERIES CROSS VALIDATION")

for fold, (train_idx, test_idx) in enumerate(tscv.split(X_reg)):
    print(
        f"Fold {fold + 1}: "
        f"train={len(train_idx)}, "
        f"test={len(test_idx)}"
    )

# Important:
#
# TimeSeriesSplit:
#   preserves temporal ordering.
#
# Never use ordinary random train_test_split for a forecasting evaluation
# when future information could leak into training.


# ============================================================
# 55. GRID SEARCH FOR ML MODELS
# ============================================================

param_grid = {
    "max_depth": [3, 5, 8],
    "min_samples_leaf": [1, 2, 5],
}

grid_search = GridSearchCV(
    estimator=RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1,
    ),
    param_grid=param_grid,
    cv=5,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1,
)

grid_search.fit(
    X_train_reg,
    y_train_reg,
)

print("\nGRID SEARCH")
print("Best parameters:", grid_search.best_params_)
print("Best CV score:", grid_search.best_score_)

# Important parameters:
#
# estimator
#   Model to tune.
#
# param_grid
#   Hyperparameter combinations to try.
#
# cv
#   Number/type of validation folds.
#
# scoring
#   Metric used to compare models.
#
# best_params_
#   Best parameter combination.


# ============================================================
# 56. COMMON FORECASTING METRICS
# ============================================================

def mape(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    non_zero = y_true != 0

    return np.mean(
        np.abs(
            (y_true[non_zero] - y_pred[non_zero])
            / y_true[non_zero]
        )
    ) * 100


def smape(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    denominator = (
        np.abs(y_true) +
        np.abs(y_pred)
    )

    mask = denominator != 0

    return np.mean(
        2 * np.abs(y_true[mask] - y_pred[mask])
        / denominator[mask]
    ) * 100


print("\nFORECAST METRICS")
print("MAE:", mean_absolute_error(y_test_ts, rf_forecast))
print("MSE:", mean_squared_error(y_test_ts, rf_forecast))
print("RMSE:", np.sqrt(mean_squared_error(y_test_ts, rf_forecast)))
print("MAPE:", mape(y_test_ts, rf_forecast))
print("sMAPE:", smape(y_test_ts, rf_forecast))

# MAE:
#   Average absolute error.
#
# MSE:
#   Average squared error.
#
# RMSE:
#   sqrt(MSE), same unit as target.
#
# MAPE:
#   Percentage error; problematic when actual values are near zero.
#
# sMAPE:
#   Symmetric percentage-style metric.


# ============================================================
# 57. FORECAST COMPARISON
# ============================================================

forecast_comparison = pd.DataFrame({
    "Actual": y_test_ts,
    "Naive": naive_prediction,
    "SeasonalNaive": seasonal_prediction,
    "HoltWinters": np.asarray(hw_forecast),
    "AR": np.asarray(ar_forecast),
    "ARIMA": np.asarray(arima_forecast),
    "SARIMA": np.asarray(sarima_forecast),
    "RF": np.asarray(rf_forecast),
    "GB": np.asarray(gb_forecast),
})

print("\nFORECAST COMPARISON")
print(forecast_comparison.head())

# Practical rule:
#
# Always compare advanced models against simple baselines.
# A complicated model should beat a reasonable baseline on the same
# evaluation setup to justify its complexity.


# ============================================================
# 58. WALK-FORWARD VALIDATION - SIMPLE PATTERN
# ============================================================

history = list(y_train_ts.values)
walk_forward_predictions = []

for actual in y_test_ts.values:

    # Example:
    # retrain/use a model using only history available so far.
    #
    # Here we use the simplest possible forecast:
    prediction = history[-1]

    walk_forward_predictions.append(prediction)

    # After prediction, actual becomes available.
    history.append(actual)

print("\nWALK-FORWARD VALIDATION")
print(
    "MAE:",
    mean_absolute_error(
        y_test_ts,
        walk_forward_predictions,
    )
)

# Key idea:
#
# At time t:
#   train/use only information available <= t
#
# Predict t+1.
# Observe actual t+1.
# Add it to history.
# Repeat.


# ============================================================
# 59. MODEL PIPELINE - SCALING + MODEL
# ============================================================

from sklearn.linear_model import Ridge

ridge_pipeline = Pipeline([
    (
        "scaler",
        StandardScaler(),
    ),
    (
        "model",
        Ridge(
            alpha=1.0,
        ),
    ),
])

ridge_pipeline.fit(
    X_train_reg,
    y_train_reg,
)

pipeline_predictions = ridge_pipeline.predict(
    X_test_reg,
)

print("\nSKLEARN PIPELINE")
print(
    "RMSE:",
    np.sqrt(
        mean_squared_error(
            y_test_reg,
            pipeline_predictions,
        )
    )
)

# Pipeline is important in production because:
#
# preprocessing -> model
#
# is stored as one object.
#
# This reduces the chance of applying different preprocessing at inference.


# ============================================================
# 60. QUICK MODEL SELECTION CHEAT SHEET
# ============================================================

"""
REGRESSION
----------
Linear Regression:
    Linear relationship / strong baseline.

Ridge:
    Linear model + L2 regularization.

Lasso:
    Linear model + L1 regularization / feature selection.

ElasticNet:
    L1 + L2.

Decision Tree:
    Nonlinear rules, easy to explain.

Random Forest:
    Bagged trees, robust nonlinear baseline.

Gradient Boosting:
    Sequential error-correcting trees.

XGBoost:
    Highly optimized gradient boosting with strong regularization controls.

LightGBM:
    Fast gradient boosting, leaf-wise tree growth.

CatBoost:
    Gradient boosting especially convenient for categorical features.

SVM:
    Margin-based model; strong in medium-sized, well-scaled datasets.

KNN:
    Distance-based; simple but expensive at prediction time.

CLASSIFICATION
--------------
Logistic Regression:
    Linear decision boundary + probabilities.

Decision Tree:
    Rule-based nonlinear classifier.

Random Forest:
    Ensemble of randomized trees.

AdaBoost:
    Sequentially focuses on hard examples.

Gradient Boosting:
    Sequential weak learners correcting residual/error.

XGBoost / LightGBM / CatBoost:
    Advanced gradient boosting.

SVM:
    Maximum-margin classification.

KNN:
    Neighbor voting.

Naive Bayes:
    Fast probabilistic baseline under conditional-independence assumption.

LDA:
    Linear discriminant classifier.

QDA:
    Quadratic decision boundaries with class-specific covariance.

CLUSTERING
----------
K-Means:
    K spherical/centroid-based clusters.

Hierarchical:
    Builds a hierarchy/dendrogram of clusters.

DBSCAN:
    Density-based clusters + noise detection.

GMM:
    Soft probabilistic clustering using Gaussian components.

DIMENSIONALITY REDUCTION
------------------------
PCA:
    Linear projection maximizing variance.

LDA:
    Supervised projection maximizing class separation.

t-SNE:
    Nonlinear visualization emphasizing local neighborhoods.

UMAP:
    Nonlinear embedding preserving local + some global structure.

ANOMALY DETECTION
-----------------
Isolation Forest:
    Isolates unusual points with random trees.

LOF:
    Detects points with unusually low local density.

One-Class SVM:
    Learns a boundary around normal data.

RECOMMENDERS
------------
Collaborative filtering:
    Uses user-item interaction patterns.

Content-based:
    Uses item features/content.

Matrix factorization:
    Learns latent user/item vectors.

Two-tower:
    User/query embedding + item/document embedding.

Retrieval + ranking:
    Retrieve candidates first, rank them second.

TIME SERIES
-----------
Naive:
    Last value.

Seasonal Naive:
    Repeat seasonal pattern.

Exponential Smoothing:
    Weighted recent observations.

AR:
    Past target values.

MA:
    Past errors.

ARMA:
    AR + MA.

ARIMA:
    AR + differencing + MA.

SARIMA:
    ARIMA + seasonal ARIMA.

SARIMAX:
    SARIMA + external variables.

VAR:
    Multiple time series predict each other.

VARX:
    VAR + external variables.

ML Forecasting:
    Convert time series into supervised features.

Deep / Foundation Forecasting:
    RNN -> LSTM/GRU -> Transformer -> PatchTST/TimesFM/Chronos.
"""


# ============================================================
# 61. MASTER ROADMAP TREE
# ============================================================

"""
ML ROADMAP WE HAVE COVERED
│
├── 1. SUPERVISED LEARNING
│   │
│   ├── Regression
│   │   ├── Linear Regression
│   │   ├── Ridge
│   │   ├── Lasso
│   │   ├── Elastic Net
│   │   └── Polynomial Regression
│   │
│   ├── Classification
│   │   ├── Logistic Regression
│   │   ├── Decision Tree
│   │   ├── Random Forest
│   │   ├── AdaBoost
│   │   ├── Gradient Boosting
│   │   ├── XGBoost
│   │   ├── LightGBM
│   │   ├── CatBoost
│   │   ├── SVM
│   │   ├── KNN
│   │   ├── Naive Bayes
│   │   ├── LDA
│   │   └── QDA
│   │
├── 2. UNSUPERVISED LEARNING
│   │
│   ├── Clustering
│   │   ├── K-Means
│   │   ├── Hierarchical
│   │   ├── DBSCAN
│   │   └── GMM
│   │
│   └── Dimensionality Reduction
│       ├── PCA
│       ├── LDA
│       ├── t-SNE
│       └── UMAP
│
├── 3. ANOMALY DETECTION
│   ├── Isolation Forest
│   ├── LOF
│   └── One-Class SVM
│
├── 4. RECOMMENDER SYSTEMS
│   ├── User-based CF
│   ├── Item-based CF
│   ├── Content-based / TF-IDF
│   ├── Matrix Factorization
│   ├── Embeddings / retrieval concept
│   └── Ranking metrics
│
└── 5. TIME SERIES
    │
    ├── Fundamentals
    ├── Lag / Rolling Features
    ├── ACF / PACF
    ├── Stationarity / ADF
    ├── Differencing
    ├── Naive / Seasonal Naive
    ├── Exponential Smoothing
    ├── AR
    ├── MA
    ├── ARMA
    ├── ARIMA
    ├── SARIMA
    ├── SARIMAX
    ├── VAR
    ├── VARX
    ├── ML Forecasting
    ├── Metrics
    ├── TimeSeriesSplit
    └── Walk-forward Validation

NEXT AFTER THIS
----------------
Deep Learning
    -> Neural Networks
    -> Backpropagation
    -> Optimization
    -> CNN
    -> RNN
    -> LSTM
    -> GRU
    -> Attention
    -> Transformers

Then:
    -> NLP
    -> Embeddings
    -> LLMs
    -> RAG
    -> Vector Databases
    -> Agents
    -> ML Systems / MLOps
    -> Production AI Systems
"""


print("\n============================================================")
print("ML / SKLEARN USAGE CHEATSHEET LOADED")
print("============================================================")
```
