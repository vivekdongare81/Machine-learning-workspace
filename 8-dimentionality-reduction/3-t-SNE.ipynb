```python
# ============================================================
# t-SNE — T-DISTRIBUTED STOCHASTIC NEIGHBOR EMBEDDING 🟢
# ============================================================
#
# DEPTH:
#   🟢 Concept + sklearn
#
# Main goal:
#
#     HIGH-DIMENSIONAL DATA
#              ↓
#             t-SNE
#              ↓
#     2D / 3D VISUALIZATION
#
#
# IMPORTANT:
#
# t-SNE is mainly a VISUALIZATION technique.
#
# It is NOT usually the first choice for preprocessing features
# before a production ML model.
#
#
# We will learn:
#
#   1. Why t-SNE?
#   2. Intuition
#   3. Local vs global structure
#   4. High-dimensional probabilities
#   5. Low-dimensional probabilities
#   6. Why Student-t distribution?
#   7. KL divergence
#   8. Gradient descent
#   9. Perplexity
#  10. Internal working
#  11. sklearn implementation
#  12. Important parameters
#  13. Visualizing MNIST-like data
#  14. PCA vs t-SNE
#  15. t-SNE vs UMAP
#  16. Limitations
#  17. Interview questions
#  18. Working tree
#
# ============================================================


# ============================================================
# 1. IMPORTS
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_digits

from sklearn.preprocessing import StandardScaler

from sklearn.decomposition import PCA

from sklearn.manifold import TSNE


# ============================================================
# 2. WHY t-SNE?
# ============================================================
#
# Suppose each data point has:
#
#     100 features
#
# We cannot directly visualize 100 dimensions.
#
#
# We want:
#
#     100D
#       ↓
#      2D
#
#
# so that similar points appear close together.
#
#
# Example:
#
# Handwritten digits:
#
#     784-dimensional images
#
# We want to visualize:
#
#     digit 0
#     digit 1
#     digit 2
#     ...
#     digit 9
#
# in a 2D plot.
#
#
# t-SNE is designed specifically for this kind of problem.
#
# ============================================================


# ============================================================
# 3. CORE INTUITION ⭐⭐⭐
# ============================================================
#
# Imagine high-dimensional points:
#
#
# High-dimensional space:
#
#       A ----- B
#
#       C
#
#                         D ----- E
#
#
# A and B are similar.
#
# D and E are similar.
#
#
# t-SNE tries to create a 2D representation where:
#
#
#        A -- B
#
#
#                       D -- E
#
#
# Similar points:
#
#     stay close.
#
#
# Dissimilar points:
#
#     are generally pushed apart.
#
#
# The MOST IMPORTANT goal is:
#
#     Preserve LOCAL neighborhoods.
#
# ============================================================


# ============================================================
# 4. LOCAL VS GLOBAL STRUCTURE ⭐⭐⭐
# ============================================================
#
# t-SNE focuses primarily on LOCAL structure.
#
#
# LOCAL:
#
#     "Which points are my nearest neighbors?"
#
#
# GLOBAL:
#
#     "What are the exact distances between far-away groups?"
#
#
# t-SNE is much better at preserving:
#
#     local neighborhoods
#
# than:
#
#     exact global geometry.
#
#
# IMPORTANT:
#
# The distance between two visually separated clusters in a
# t-SNE plot should NOT automatically be interpreted as their
# true distance in the original feature space.
#
# ============================================================


# ============================================================
# 5. WHY NOT JUST USE PCA?
# ============================================================
#
# PCA tries to find linear directions that preserve variance.
#
#
# t-SNE tries to preserve neighborhood relationships.
#
#
# Example:
#
# Data may lie on a complex nonlinear manifold:
#
#
#             ● ●
#           ●     ●
#          ●       ●
#           ●     ●
#             ● ●
#
#
# PCA may not unfold this nonlinear structure.
#
# t-SNE can often create a visualization where local groups
# become much more apparent.
#
# ============================================================


# ============================================================
# 6. THE KEY IDEA OF t-SNE ⭐⭐⭐
# ============================================================
#
# t-SNE does NOT directly try to preserve raw distances.
#
# Instead:
#
#     DISTANCES
#         ↓
#     PROBABILITIES
#
#
# It creates:
#
#     probabilities of neighboring points.
#
#
# Then it tries to make the probabilities in the low-dimensional
# space resemble the probabilities from the original space.
#
#
# So:
#
#     High-dimensional relationships
#                 ↓
#             probabilities
#                 ↓
#              t-SNE
#                 ↓
#             2D points
#                 ↓
#          low-dimensional
#          probabilities
#
# ============================================================


# ============================================================
# 7. HIGH-DIMENSIONAL SIMILARITY ⭐⭐⭐
# ============================================================
#
# For two points x_i and x_j, t-SNE converts their distance
# into a conditional probability.
#
#
# Conceptually:
#
#
#     p(j | i)
#
# represents:
#
#     "How likely is j to be a neighbor of i?"
#
#
# Nearby points:
#
#     high probability
#
#
# Far-away points:
#
#     low probability
#
#
# A Gaussian distribution is used to convert distances into
# probabilities.
#
# ============================================================


# ============================================================
# 8. HIGH-DIMENSIONAL CONDITIONAL PROBABILITY
# ============================================================
#
# Simplified formula:
#
#
#                  exp(
#                    -||x_i - x_j||²
#                    ----------------
#                    2σ_i²
#                  )
#
# p(j|i) = ---------------------------------
#
#          Σ_{k != i} exp(
#                    -||x_i - x_k||²
#                    ----------------
#                    2σ_i²
#                  )
#
#
# where:
#
#     σ_i = bandwidth for point i.
#
#
# The important intuition:
#
#     smaller distance
#          ↓
#     larger probability
#
#     larger distance
#          ↓
#     smaller probability
#
# ============================================================


# ============================================================
# 9. WHY DIFFERENT σ FOR EACH POINT?
# ============================================================
#
# Data density can vary.
#
#
# Dense region:
#
#     Many neighbors are close.
#
#
# Sparse region:
#
#     Neighbors are farther away.
#
#
# Therefore t-SNE adapts the Gaussian bandwidth σ_i for each
# point.
#
#
# Perplexity controls this neighborhood scale.
#
# ============================================================


# ============================================================
# 10. SYMMETRIZATION
# ============================================================
#
# The conditional probability:
#
#     p(j|i)
#
# is not necessarily equal to:
#
#     p(i|j)
#
#
# t-SNE creates a symmetric similarity:
#
#
#     p_ij = (p(j|i) + p(i|j)) / (2N)
#
#
# where N is the number of points.
#
#
# Now we have a probability distribution representing
# similarities in the original space.
#
# ============================================================


# ============================================================
# 11. LOW-DIMENSIONAL SPACE
# ============================================================
#
# Now randomly initialize low-dimensional points:
#
#
#     y_1, y_2, ..., y_N
#
#
# Usually:
#
#     2D
#
# or:
#
#     3D
#
#
# t-SNE calculates another similarity distribution:
#
#     q_ij
#
# in the low-dimensional space.
#
# ============================================================


# ============================================================
# 12. WHY STUDENT-t DISTRIBUTION? ⭐⭐⭐
# ============================================================
#
# This is one of the most important ideas in t-SNE.
#
#
# In high dimensions:
#
#     moderate/far distances can behave differently from
#     low-dimensional distances.
#
#
# A Gaussian distribution in the low-dimensional space can
# cause the "crowding problem".
#
#
# t-SNE therefore uses a:
#
#     Student-t distribution
#
# with one degree of freedom.
#
#
# This is essentially a Cauchy-like heavy-tailed distribution.
#
#
# Simplified similarity:
#
#
#                 (1 + ||y_i-y_j||²)^(-1)
#
# q_ij = ----------------------------------------
#             Σ_{k != l}
#                 (1 + ||y_k-y_l||²)^(-1)
#
#
# Heavy tails give distant points more room.
#
# ============================================================


# ============================================================
# 13. THE CROWDING PROBLEM ⭐⭐⭐
# ============================================================
#
# Imagine many points in high dimensions.
#
# There can be many directions in which points can spread out.
#
# In 2D there is much less room.
#
#
# If we use a Gaussian distribution in 2D:
#
#     many moderately distant points can get crowded together.
#
#
# Student-t has heavier tails:
#
#     distant points receive more probability mass.
#
#
# This helps separate points in the low-dimensional map.
#
# ============================================================


# ============================================================
# 14. KL DIVERGENCE ⭐⭐⭐
# ============================================================
#
# Now we have:
#
#
#     P = similarities in high-dimensional space
#
#     Q = similarities in low-dimensional space
#
#
# t-SNE minimizes:
#
#
#     KL(P || Q)
#
#
# Formula:
#
#
#              p_ij
# KL(P||Q) = Σ p_ij log( ---- )
#              i,j  q_ij
#
#
# Intuition:
#
#     Make Q resemble P.
#
#
# If two points are close in high-dimensional space:
#
#     p_ij is large.
#
# Therefore t-SNE strongly tries to keep them close in the
# low-dimensional space.
#
# ============================================================


# ============================================================
# 15. WHY KL DIRECTION MATTERS ⭐⭐
# ============================================================
#
# t-SNE minimizes:
#
#     KL(P || Q)
#
#
# not:
#
#     KL(Q || P)
#
#
# This makes missing a strong high-dimensional neighbor
# relatively expensive.
#
#
# Therefore t-SNE strongly focuses on preserving local
# neighborhoods.
#
# ============================================================


# ============================================================
# 16. OPTIMIZATION ⭐⭐⭐
# ============================================================
#
# We start with random low-dimensional positions.
#
#
# Then:
#
#
#     Calculate Q
#          ↓
#     Calculate KL divergence
#          ↓
#     Calculate gradient
#          ↓
#     Move points
#          ↓
#     Recalculate Q
#          ↓
#     Repeat
#
#
# This is gradient-based optimization.
#
#
# Eventually we obtain a 2D embedding that approximately
# preserves the high-dimensional neighborhood probabilities.
#
# ============================================================


# ============================================================
# 17. COMPLETE t-SNE INTERNAL FLOW ⭐⭐⭐
# ============================================================
#
#
#                  HIGH-D DATA
#                       │
#                       ▼
#                Calculate distances
#                       │
#                       ▼
#           Convert distances → P_ij
#                       │
#                       ▼
#                Symmetrize P
#                       │
#                       ▼
#              Initialize 2D points
#                       │
#                       ▼
#           Calculate low-D Q_ij
#              using Student-t
#                       │
#                       ▼
#             Calculate KL(P || Q)
#                       │
#                       ▼
#                  Gradient
#                       │
#                       ▼
#              Update 2D points
#                       │
#                       ▼
#                 Repeat
#                       │
#                       ▼
#                  Final 2D map
#
# ============================================================


# ============================================================
# 18. LOAD A HIGH-DIMENSIONAL DATASET
# ============================================================
#
# Digits dataset:
#
#     8 x 8 image
#
#     64 features
#
#
# Each image becomes:
#
#     64-dimensional vector.
#
# ============================================================

digits = load_digits()

X = digits.data

y = digits.target

print(
    "Original shape:",
    X.shape
)


# ============================================================
# 19. SCALE DATA
# ============================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    X
)


# ============================================================
# 20. BASIC t-SNE
# ============================================================
#
# n_components=2:
#
#     Create a 2D representation.
#
#
# perplexity:
#
#     Controls effective neighborhood size.
#
#
# random_state:
#
#     Makes the experiment reproducible.
#
# ============================================================

tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

X_tsne = tsne.fit_transform(
    X_scaled
)

print(
    "t-SNE shape:",
    X_tsne.shape
)


# ============================================================
# 21. VISUALIZE t-SNE
# ============================================================

plt.figure(figsize=(10, 8))

plt.scatter(
    X_tsne[:, 0],
    X_tsne[:, 1],
    c=y,
    s=10
)

plt.xlabel(
    "t-SNE 1"
)

plt.ylabel(
    "t-SNE 2"
)

plt.title(
    "Digits Dataset — t-SNE"
)

plt.show()


# ============================================================
# 22. WHAT ARE WE LOOKING FOR?
# ============================================================
#
# Each point:
#
#     one image.
#
#
# Points with similar digit patterns should often appear
# close together.
#
#
# For example:
#
#     similar "0" images
#           ↓
#        local group
#
#
#     similar "1" images
#           ↓
#        local group
#
#
# The colors are the TRUE labels used only to inspect the
# resulting visualization.
#
#
# IMPORTANT:
#
# Standard t-SNE itself does not need y.
#
# ============================================================


# ============================================================
# 23. PERPLEXITY ⭐⭐⭐
# ============================================================
#
# Perplexity is one of the most important t-SNE parameters.
#
#
# Rough intuition:
#
#     Effective number of neighbors considered.
#
#
# Common values:
#
#     5
#     10
#     30
#     50
#
#
# Small perplexity:
#
#     Focus more strongly on very local neighborhoods.
#
#
# Large perplexity:
#
#     Consider broader neighborhoods.
#
#
# It is NOT exactly "number of neighbors".
#
# Think:
#
#     effective neighborhood size.
#
# ============================================================


# ============================================================
# 24. PERPLEXITY EXPERIMENT
# ============================================================
#
# Compare several values.
#
# ============================================================

perplexities = [
    5,
    30,
    50
]

for p in perplexities:

    model = TSNE(
        n_components=2,
        perplexity=p,
        random_state=42
    )

    embedding = model.fit_transform(
        X_scaled
    )

    plt.figure(figsize=(8, 6))

    plt.scatter(
        embedding[:, 0],
        embedding[:, 1],
        c=y,
        s=10
    )

    plt.xlabel(
        "t-SNE 1"
    )

    plt.ylabel(
        "t-SNE 2"
    )

    plt.title(
        f"t-SNE — perplexity={p}"
    )

    plt.show()


# ============================================================
# 25. HOW PERPLEXITY CHANGES THE RESULT
# ============================================================
#
#
# Small perplexity:
#
#     Smaller neighborhoods
#          ↓
#     Strong local focus
#
#
# Large perplexity:
#
#     Larger neighborhoods
#          ↓
#     More global context
#
#
# There is no universally correct perplexity.
#
# It depends on:
#
#     dataset size
#     density
#     structure
#
# ============================================================


# ============================================================
# 26. LEARNING RATE ⭐⭐⭐
# ============================================================
#
# learning_rate controls how strongly the embedding positions
# are updated during optimization.
#
#
# Too small:
#
#     optimization can move slowly.
#
#
# Too large:
#
#     optimization can become unstable or produce poor
#     embeddings.
#
#
# In modern sklearn versions, learning_rate="auto" is commonly
# used as a default.
#
# ============================================================


# ============================================================
# 27. EARLY EXAGGERATION ⭐⭐
# ============================================================
#
# During early optimization, t-SNE temporarily increases the
# importance of attractive forces between points.
#
#
# This is called:
#
#     Early Exaggeration
#
#
# Purpose:
#
#     Help form meaningful clusters early in optimization.
#
#
# Parameter:
#
#     early_exaggeration
#
# ============================================================


# ============================================================
# 28. ITERATIONS ⭐⭐
# ============================================================
#
# t-SNE is iterative.
#
#
# More optimization iterations:
#
#     potentially better convergence
#     more computation
#
#
# Current sklearn APIs use:
#
#     max_iter
#
# for the maximum number of optimization iterations.
#
# ============================================================


# ============================================================
# 29. INIT ⭐⭐
# ============================================================
#
# Initial low-dimensional positions can affect the final result.
#
#
# Common choices:
#
#     "random"
#     "pca"
#
#
# PCA initialization can sometimes provide a more structured
# starting point than random initialization.
#
# ============================================================


# ============================================================
# 30. RANDOMNESS ⭐⭐⭐
# ============================================================
#
# t-SNE can produce different embeddings across runs because
# the optimization can depend on initialization and stochastic
# behavior.
#
#
# Therefore:
#
#     random_state=42
#
# is useful when reproducing experiments.
#
#
# IMPORTANT:
#
# Different t-SNE runs can produce different rotations,
# reflections, or layouts while representing similar local
# structures.
#
# ============================================================


# ============================================================
# 31. IMPORTANT t-SNE PARAMETERS
# ============================================================
#
#
# TSNE(
#
#     n_components=2,
#     perplexity=30.0,
#     early_exaggeration=12.0,
#     learning_rate="auto",
#     max_iter=1000,
#     n_iter_without_progress=300,
#     min_grad_norm=1e-7,
#     metric="euclidean",
#     init="pca",
#     random_state=None
#
# )
#
#
# ------------------------------------------------------------
# n_components ⭐⭐⭐
# ------------------------------------------------------------
#
# Output dimensions.
#
# Usually:
#
#     2
#
# or:
#
#     3
#
# for visualization.
#
#
# ------------------------------------------------------------
# perplexity ⭐⭐⭐
# ------------------------------------------------------------
#
# Effective neighborhood size.
#
#
# ------------------------------------------------------------
# learning_rate ⭐⭐⭐
# ------------------------------------------------------------
#
# Controls optimization step size.
#
#
# ------------------------------------------------------------
# early_exaggeration ⭐⭐
# ------------------------------------------------------------
#
# Strengthens attractive forces during early optimization.
#
#
# ------------------------------------------------------------
# max_iter ⭐⭐
# ------------------------------------------------------------
#
# Maximum optimization iterations.
#
#
# ------------------------------------------------------------
# metric ⭐⭐
# ------------------------------------------------------------
#
# Distance metric used to calculate similarities in the
# original space.
#
#
# ------------------------------------------------------------
# init
# ------------------------------------------------------------
#
# Initialization strategy:
#
#     "random"
#     "pca"
#
#
# ------------------------------------------------------------
# random_state
# ------------------------------------------------------------
#
# Reproducibility.
#
# ============================================================


# ============================================================
# 32. PCA BEFORE t-SNE ⭐⭐⭐
# ============================================================
#
# A common workflow for high-dimensional data:
#
#
#     Original data
#          ↓
#         PCA
#          ↓
#    30-50 dimensions
#          ↓
#        t-SNE
#          ↓
#         2D
#
#
# Why?
#
# PCA can remove some noise/redundancy and make t-SNE
# computationally easier.
#
#
# IMPORTANT:
#
# The exact number of PCA dimensions is dataset-dependent.
#
# ============================================================


# ============================================================
# 33. PCA → t-SNE PIPELINE
# ============================================================

pca = PCA(
    n_components=30,
    random_state=42
)

X_pca = pca.fit_transform(
    X_scaled
)

print(
    "\nAfter PCA:",
    X_pca.shape
)


tsne_after_pca = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

X_tsne_pca = tsne_after_pca.fit_transform(
    X_pca
)


# ============================================================
# 34. VISUALIZE PCA → t-SNE
# ============================================================

plt.figure(figsize=(10, 8))

plt.scatter(
    X_tsne_pca[:, 0],
    X_tsne_pca[:, 1],
    c=y,
    s=10
)

plt.xlabel(
    "t-SNE 1"
)

plt.ylabel(
    "t-SNE 2"
)

plt.title(
    "PCA → t-SNE"
)

plt.show()


# ============================================================
# 35. IMPORTANT: t-SNE IS NOT A STANDARD FEATURE EXTRACTOR
# ============================================================
#
# Avoid thinking:
#
#     t-SNE → 2 features → train classifier
#
# as the default workflow.
#
#
# Why?
#
# t-SNE is optimized mainly for visualization and neighborhood
# preservation.
#
# The embedding can depend on:
#
#     initialization
#     perplexity
#     optimization
#     random state
#
#
# For production feature reduction, methods such as:
#
#     PCA
#     learned neural embeddings
#     UMAP (depending on use case)
#
# may be more appropriate.
#
# ============================================================


# ============================================================
# 36. t-SNE DOES NOT PRESERVE GLOBAL DISTANCES ⭐⭐⭐
# ============================================================
#
# This is extremely important.
#
#
# Suppose t-SNE produces:
#
#
#       Cluster A
#          ●●●
#          ●●●
#
#
#
#
#                       Cluster B
#                         ●●●
#                         ●●●
#
#
# You should NOT conclude:
#
#     "A is exactly twice as far from B as C is from D."
#
#
# t-SNE primarily preserves local neighborhood relationships.
#
#
# Cluster size, inter-cluster distance, and geometry can be
# visually misleading.
#
# ============================================================


# ============================================================
# 37. CLUSTER SIZE WARNING ⭐⭐
# ============================================================
#
# A large-looking cluster in a t-SNE plot does not necessarily
# mean:
#
#     "This class occupies more volume in the original space."
#
#
# t-SNE is constructing a visualization.
#
# It can distort:
#
#     global distances
#     cluster sizes
#     density
#     geometry
#
#
# Therefore interpret the visualization carefully.
#
# ============================================================


# ============================================================
# 38. t-SNE VS PCA ⭐⭐⭐
# ============================================================
#
#
# PCA:
#
#     Linear
#
#     Unsupervised
#
#     Maximizes variance
#
#     Fast
#
#     Useful for preprocessing
#
#     Preserves global linear structure better
#
#
# t-SNE:
#
#     Nonlinear
#
#     Unsupervised
#
#     Preserves local neighborhoods
#
#     More computationally expensive
#
#     Primarily visualization
#
#
# Mental model:
#
#
# PCA:
#
#     "Find the best linear directions."
#
#
# t-SNE:
#
#     "Put similar neighbors close together in 2D."
#
# ============================================================


# ============================================================
# 39. t-SNE VS UMAP ⭐⭐⭐
# ============================================================
#
#
# t-SNE:
#
#     Strong local structure.
#
#     Mainly visualization.
#
#     Can be computationally expensive.
#
#
# UMAP:
#
#     Nonlinear manifold learning.
#
#     Preserves local structure while often retaining more
#     global organization than t-SNE.
#
#     Usually faster on larger datasets.
#
#
# IMPORTANT:
#
# These are tendencies, not guarantees for every dataset.
#
# ============================================================


# ============================================================
# 40. t-SNE VS LDA ⭐⭐⭐
# ============================================================
#
#
# t-SNE:
#
#     Unsupervised.
#
#     Does not require y.
#
#
# LDA:
#
#     Supervised.
#
#     Requires y.
#
#
# t-SNE:
#
#     Neighborhood visualization.
#
#
# LDA:
#
#     Class-discriminative projection.
#
# ============================================================


# ============================================================
# 41. ADVANTAGES
# ============================================================
#
# 1. Excellent for visualizing high-dimensional data.
#
# 2. Captures nonlinear local structure.
#
# 3. Can reveal groups that are difficult to see with PCA.
#
# 4. Useful for exploratory data analysis.
#
# 5. Useful for inspecting learned embeddings.
#
# ============================================================


# ============================================================
# 42. LIMITATIONS ⭐⭐⭐
# ============================================================
#
# 1. Computationally expensive.
#
# 2. Sensitive to hyperparameters.
#
# 3. Can produce different layouts across runs.
#
# 4. Primarily designed for visualization.
#
# 5. Does not preserve global distances reliably.
#
# 6. Interpretation of cluster distances can be misleading.
#
# 7. Scaling to very large datasets can be challenging.
#
# ============================================================


# ============================================================
# 43. WHEN TO USE t-SNE
# ============================================================
#
# Good use cases:
#
#     ✓ Visualizing image embeddings
#     ✓ Visualizing text embeddings
#     ✓ Exploring clusters
#     ✓ Understanding learned representations
#     ✓ Exploratory data analysis
#
#
# Less appropriate as the default:
#
#     ✗ Production feature preprocessing
#     ✗ Preserving exact global distances
#     ✗ Interpreting cluster sizes quantitatively
#
# ============================================================


# ============================================================
# 44. INTERVIEW QUESTIONS ⭐⭐⭐
# ============================================================
#
#
# Q1. What is t-SNE?
#
# Answer:
#
# t-SNE is a nonlinear dimensionality reduction technique
# primarily used to visualize high-dimensional data in 2D
# or 3D while preserving local neighborhood relationships.
#
#
# ------------------------------------------------------------
#
# Q2. What does t-SNE preserve?
#
# Answer:
#
# Primarily local neighborhood structure.
#
#
# ------------------------------------------------------------
#
# Q3. Does t-SNE preserve global distances?
#
# Answer:
#
# Not reliably. The method is designed mainly to preserve
# local relationships, so distances between separate clusters
# should not be interpreted as exact global distances.
#
#
# ------------------------------------------------------------
#
# Q4. Why does t-SNE use a Student-t distribution?
#
# Answer:
#
# The heavy tails provide more room for moderately distant
# points in the low-dimensional space and help address the
# crowding problem.
#
#
# ------------------------------------------------------------
#
# Q5. What is perplexity?
#
# Answer:
#
# Perplexity controls the effective neighborhood size used
# when determining high-dimensional similarities.
#
#
# ------------------------------------------------------------
#
# Q6. What happens if perplexity is too small?
#
# Answer:
#
# The embedding focuses strongly on very local neighborhoods
# and may produce overly fragmented local structures.
#
#
# ------------------------------------------------------------
#
# Q7. What happens if perplexity is too large?
#
# Answer:
#
# Larger neighborhoods influence the embedding, potentially
# changing the local structure and making small groups less
# distinct.
#
#
# ------------------------------------------------------------
#
# Q8. What is KL divergence doing in t-SNE?
#
# Answer:
#
# It measures the difference between the high-dimensional
# similarity distribution P and the low-dimensional similarity
# distribution Q. t-SNE minimizes KL(P || Q).
#
#
# ------------------------------------------------------------
#
# Q9. Does t-SNE use labels?
#
# Answer:
#
# No. Standard t-SNE is unsupervised. Labels can be used
# afterward only to color or inspect the visualization.
#
#
# ------------------------------------------------------------
#
# Q10. Why use PCA before t-SNE?
#
# Answer:
#
# For high-dimensional data, PCA can reduce noise and
# dimensionality before t-SNE, making the subsequent
# optimization more manageable.
#
#
# ------------------------------------------------------------
#
# Q11. PCA vs t-SNE?
#
# Answer:
#
# PCA is linear and variance-based, while t-SNE is nonlinear
# and primarily preserves local neighborhoods for visualization.
#
#
# ------------------------------------------------------------
#
# Q12. Can t-SNE be used for classification?
#
# Answer:
#
# It can technically produce an embedding that could be used
# as input to another model, but that is generally not its
# primary purpose. t-SNE is mainly used for visualization.
#
#
# ============================================================
# 45. 30-SECOND INTERVIEW EXPLANATION ⭐⭐⭐
# ============================================================
#
# t-SNE is a nonlinear dimensionality reduction technique
# mainly used for visualizing high-dimensional data.
#
# It converts pairwise similarities in the original space
# into probabilities and then finds a low-dimensional
# representation whose probability distribution is similar.
#
# It uses a Student-t distribution in the low-dimensional
# space and minimizes KL divergence between the two
# distributions.
#
# Its main strength is preserving local neighborhoods, not
# exact global distances.
#
# ============================================================


# ============================================================
# 46. MEMORY CHEATSHEET
# ============================================================
#
#
# t-SNE
#   │
#   ├── Nonlinear
#   │
#   ├── Unsupervised
#   │
#   ├── Mainly visualization
#   │
#   ├── Preserve local neighborhoods
#   │
#   ├── High-D similarities → P
#   │
#   ├── Low-D similarities → Q
#   │
#   ├── Student-t in low-D
#   │
#   ├── KL(P || Q)
#   │
#   └── Gradient descent
#
#
# MOST IMPORTANT PARAMETER:
#
#     perplexity
#          ↓
#     effective neighborhood size
#
#
# ============================================================
# 47. INTERNAL WORKING TREE ⭐⭐⭐
# ============================================================
#
#
#                       t-SNE
#                         │
#                         ▼
#                 High-dimensional X
#                         │
#                         ▼
#                   Pairwise distances
#                         │
#                         ▼
#             Convert distances → P_ij
#                         │
#                         ▼
#                   Symmetrize P
#                         │
#                         ▼
#                Initialize low-D Y
#                         │
#                         ▼
#              Calculate q_ij using
#                Student-t distribution
#                         │
#                         ▼
#                  KL(P || Q)
#                         │
#                         ▼
#                     Gradient
#                         │
#                         ▼
#                 Update Y positions
#                         │
#                         ▼
#                     Repeat
#                         │
#                         ▼
#                    2D embedding
#
#
# ============================================================
#
# ONE-LINE MENTAL MODEL:
#
#     "Turn high-dimensional neighbor relationships into
#      probabilities, then move points in 2D until their
#      neighborhood probabilities look similar."
#
# ============================================================
```
