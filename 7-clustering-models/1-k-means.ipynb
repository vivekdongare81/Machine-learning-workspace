```python
# ============================================================
# K-MEANS CLUSTERING 🟡
# ============================================================
#
# DEPTH:
#   🟡 Medium
#
# No scratch implementation required.
#
# Learn:
#   1. Why K-Means?
#   2. Intuition
#   3. Centroids
#   4. Assignment step
#   5. Update step
#   6. Objective function / inertia
#   7. Internal working
#   8. sklearn implementation
#   9. Important parameters
#  10. Choosing K
#  11. Elbow method
#  12. Silhouette score
#  13. Initialization
#  14. K-Means++
#  15. Advantages / disadvantages
#  16. K-Means vs GMM
#  17. Interview questions
#  18. 30-second explanation
#  19. Working tree
#
# ============================================================


# ============================================================
# 1. IMPORTS
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs

from sklearn.cluster import KMeans

from sklearn.metrics import silhouette_score


# ============================================================
# 2. WHY K-MEANS?
# ============================================================
#
# K-Means is an UNSUPERVISED clustering algorithm.
#
# That means:
#
#     We do NOT have target labels.
#
#
# Example:
#
# Customer data:
#
#     Age
#     Income
#     Spending
#
#
# We don't know customer groups beforehand.
#
#
# K-Means can discover groups such as:
#
#     Cluster 1 → low spending
#     Cluster 2 → medium spending
#     Cluster 3 → high spending
#
#
# The algorithm groups similar points together.
#
# ============================================================


# ============================================================
# 3. CORE INTUITION ⭐⭐⭐
# ============================================================
#
# Imagine this dataset:
#
#
#       ● ● ●
#      ● ● ●
#
#
#                    ○ ○ ○
#                   ○ ○ ○
#
#
#                            ▲ ▲
#                           ▲ ▲ ▲
#
#
# We want:
#
#     Cluster 1 → ●
#     Cluster 2 → ○
#     Cluster 3 → ▲
#
#
# K-Means does this using CENTROIDS.
#
# A centroid is the center of a cluster.
#
# ============================================================


# ============================================================
# 4. WHAT IS K?
# ============================================================
#
# K = number of clusters we want.
#
#
# Example:
#
#     K = 3
#
# means:
#
#     Find 3 clusters.
#
#
# IMPORTANT:
#
# K is NOT automatically discovered by K-Means.
#
# We usually need to choose K using techniques such as:
#
#     Elbow method
#     Silhouette score
#     Domain knowledge
#
# ============================================================


# ============================================================
# 5. WHAT IS A CENTROID? ⭐⭐⭐
# ============================================================
#
# A centroid is the mean position of all points assigned
# to a cluster.
#
#
# Example:
#
# Points:
#
#     (1,1)
#     (2,2)
#     (3,3)
#
#
# Centroid:
#
#     x = (1+2+3)/3 = 2
#
#     y = (1+2+3)/3 = 2
#
#
# Centroid:
#
#     (2,2)
#
#
# Therefore:
#
#     Centroid = mean of points in that cluster.
#
# ============================================================


# ============================================================
# 6. K-MEANS INTERNAL WORKING ⭐⭐⭐
# ============================================================
#
# K-Means repeatedly performs TWO main steps:
#
#
#     1. Assignment step
#
#     2. Update step
#
#
# --------------------------------
# Assignment Step
# --------------------------------
#
# Assign every point to the nearest centroid.
#
#
# Point X:
#
#     distance to C1 = 2
#     distance to C2 = 5
#     distance to C3 = 3
#
#
# Assign:
#
#     X → Cluster 1
#
#
# --------------------------------
# Update Step
# --------------------------------
#
# Recalculate each centroid as the mean of its assigned
# points.
#
#
# Then repeat.
#
# ============================================================


# ============================================================
# 7. COMPLETE K-MEANS FLOW ⭐⭐⭐
# ============================================================
#
#
#                 Choose K
#                    │
#                    ▼
#            Initialize centroids
#                    │
#                    ▼
#             Assignment Step
#                    │
#                    ▼
#          Assign points to nearest
#               centroid
#                    │
#                    ▼
#               Update Step
#                    │
#                    ▼
#        Recalculate cluster means
#                    │
#                    ▼
#               Converged?
#                /       \
#              NO         YES
#              │           │
#              └── repeat  ▼
#                       Final clusters
#
# ============================================================


# ============================================================
# 8. SMALL NUMERICAL EXAMPLE ⭐⭐⭐
# ============================================================
#
# Suppose:
#
# K = 2
#
#
# Points:
#
#     A = (1,1)
#     B = (2,2)
#     C = (3,1)
#     D = (8,8)
#     E = (9,9)
#
#
# Initial centroids:
#
#     C1 = (1,1)
#     C2 = (8,8)
#
#
# Assignment:
#
# A → C1
# B → C1
# C → C1
# D → C2
# E → C2
#
#
# Update:
#
# New C1:
#
#     mean of A,B,C
#
#     x = (1+2+3)/3 = 2
#
#     y = (1+2+1)/3 = 1.33
#
#
# C1 = (2, 1.33)
#
#
# New C2:
#
#     mean of D,E
#
#     x = (8+9)/2 = 8.5
#
#     y = (8+9)/2 = 8.5
#
#
# C2 = (8.5,8.5)
#
#
# Repeat until assignments stop changing.
#
# ============================================================


# ============================================================
# 9. DISTANCE USED BY K-MEANS
# ============================================================
#
# Standard K-Means uses Euclidean distance.
#
#
# Distance between:
#
#     X = (x1,x2)
#     C = (c1,c2)
#
#
# d(X,C) =
#
# sqrt(
#     (x1-c1)^2 +
#     (x2-c2)^2
# )
#
#
# K-Means assigns the point to the centroid with the
# smallest distance.
#
# ============================================================


# ============================================================
# 10. K-MEANS OBJECTIVE FUNCTION ⭐⭐⭐
# ============================================================
#
# K-Means tries to minimize:
#
#
#       K
#      ---
# J =  \   Σ ||x_i - μ_k||²
#      /
#      ---
#
#
# In simple words:
#
#     Minimize the squared distance between every point
#     and its assigned centroid.
#
#
# This is called:
#
#     Within-Cluster Sum of Squares
#
# or:
#
#     Inertia
#
#
# sklearn calls this:
#
#     inertia_
#
# ============================================================


# ============================================================
# 11. CREATE DATASET
# ============================================================

X, true_labels = make_blobs(
    n_samples=600,
    centers=[
        [-4, -3],
        [0, 4],
        [5, -1]
    ],
    cluster_std=[
        1.0,
        1.2,
        1.0
    ],
    random_state=42
)

print(
    "Data shape:",
    X.shape
)


# ============================================================
# 12. VISUALIZE DATA
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    X[:, 0],
    X[:, 1]
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.title(
    "K-Means Dataset"
)

plt.show()


# ============================================================
# 13. BASIC K-MEANS
# ============================================================

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

kmeans.fit(X)


# ============================================================
# 14. GET CLUSTER LABELS
# ============================================================

labels = kmeans.labels_

print(
    "\nFirst 20 cluster labels:"
)

print(
    labels[:20]
)


# ============================================================
# 15. PREDICT CLUSTERS
# ============================================================
#
# predict() assigns new points to the nearest centroid.
# ============================================================

new_points = np.array([
    [-4, -3],
    [0, 4],
    [5, -1]
])

predicted_clusters = kmeans.predict(
    new_points
)

print(
    "\nNew point cluster assignments:"
)

print(
    predicted_clusters
)


# ============================================================
# 16. CLUSTER CENTERS ⭐⭐⭐
# ============================================================
#
# cluster_centers_ contains the final centroids.
# ============================================================

print(
    "\nCluster centers:"
)

print(
    kmeans.cluster_centers_
)


# ============================================================
# 17. VISUALIZE CLUSTERS + CENTROIDS
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    s=30
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker="X",
    s=200,
    label="Centroids"
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.title(
    "K-Means Clusters and Centroids"
)

plt.legend()

plt.show()


# ============================================================
# 18. INERTIA ⭐⭐⭐
# ============================================================
#
# inertia_ =
#
# Sum of squared distances between each point and its
# assigned centroid.
#
#
# Lower inertia:
#
#     points are closer to their centroids.
#
#
# BUT:
#
# Increasing K always tends to decrease inertia.
#
# Therefore we cannot simply choose the K with the
# lowest inertia.
#
# ============================================================

print(
    "\nK-Means inertia:",
    kmeans.inertia_
)


# ============================================================
# 19. ELBOW METHOD ⭐⭐⭐
# ============================================================
#
# Train K-Means with different K values.
#
# Calculate inertia for each K.
#
# Plot:
#
#     K vs inertia
#
#
# We look for an "elbow":
#
#
# Inertia
#   │
#   │\
#   │ \
#   │  \
#   │   \__
#   │      \__
#   │         \__
#   └──────────────── K
#
#
# Before elbow:
#
#     Increasing K gives large improvement.
#
#
# After elbow:
#
#     Increasing K gives smaller improvement.
#
# ============================================================


k_values = range(1, 11)

inertias = []

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X)

    inertias.append(
        model.inertia_
    )


plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    inertias,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.title(
    "Elbow Method"
)

plt.show()


# ============================================================
# 20. SILHOUETTE SCORE ⭐⭐⭐
# ============================================================
#
# Silhouette score measures:
#
#     How similar a point is to its own cluster
#
# compared with:
#
#     The nearest other cluster.
#
#
# Range:
#
#     -1 to +1
#
#
# Higher generally means better-separated clusters.
#
#
# Rough intuition:
#
#     Close to +1
#         → well separated
#
#     Around 0
#         → overlapping clusters
#
#     Negative
#         → possible wrong assignment
#
# ============================================================


silhouette_scores = []

k_values = range(2, 11)

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    cluster_labels = model.fit_predict(X)

    score = silhouette_score(
        X,
        cluster_labels
    )

    silhouette_scores.append(
        score
    )


plt.figure(figsize=(8, 5))

plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")

plt.title(
    "Silhouette Score vs K"
)

plt.show()


# ============================================================
# 21. CHOOSE K USING SILHOUETTE SCORE
# ============================================================

best_k = list(k_values)[
    np.argmax(silhouette_scores)
]

print(
    "\nBest K according to silhouette score:",
    best_k
)


# ============================================================
# 22. FINAL MODEL
# ============================================================

final_kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

final_labels = final_kmeans.fit_predict(X)

print(
    "\nFinal cluster centers:"
)

print(
    final_kmeans.cluster_centers_
)


# ============================================================
# 23. K-MEANS++ INITIALIZATION ⭐⭐⭐
# ============================================================
#
# Bad initialization can cause poor clustering.
#
#
# sklearn's default initialization is:
#
#     init="k-means++"
#
#
# K-Means++ chooses initial centroids in a smarter way
# so that they are spread out.
#
#
# Basic idea:
#
#     First centroid
#          ↓
#     Choose future centroids with preference for points
#     far away from existing centroids
#
#
# This usually gives better starting positions than
# completely random initialization.
#
# ============================================================


# ============================================================
# 24. K-MEANS++ EXPLICITLY
# ============================================================

kmeans_plus = KMeans(
    n_clusters=3,
    init="k-means++",
    random_state=42,
    n_init=10
)

kmeans_plus.fit(X)

print(
    "\nK-Means++ inertia:",
    kmeans_plus.inertia_
)


# ============================================================
# 25. RANDOM INITIALIZATION
# ============================================================
#
# We can also use:
#
#     init="random"
#
# This can be more sensitive to initialization.
#
# ============================================================

kmeans_random = KMeans(
    n_clusters=3,
    init="random",
    random_state=42,
    n_init=10
)

kmeans_random.fit(X)

print(
    "\nRandom initialization inertia:",
    kmeans_random.inertia_
)


# ============================================================
# 26. N_INIT ⭐⭐⭐
# ============================================================
#
# K-Means can produce different results depending on
# initial centroid positions.
#
#
# n_init=10
#
# means:
#
#     Run K-Means multiple times
#     with different initializations.
#
#
# Then:
#
#     Choose the solution with the lowest inertia.
#
#
# This makes the result more robust.
#
# ============================================================


# ============================================================
# 27. IMPORTANT K-MEANS PARAMETERS
# ============================================================
#
#
# ------------------------------------------------------------
# n_clusters ⭐⭐⭐
# ------------------------------------------------------------
#
# Number of clusters K.
#
#
# ------------------------------------------------------------
# init ⭐⭐⭐
# ------------------------------------------------------------
#
# Initial centroid strategy.
#
# Common:
#
#     "k-means++"
#     "random"
#
#
# ------------------------------------------------------------
# n_init ⭐⭐⭐
# ------------------------------------------------------------
#
# Number of independent initializations.
#
#
# More:
#
#     potentially better solution
#     more computation
#
#
# ------------------------------------------------------------
# max_iter
# ------------------------------------------------------------
#
# Maximum number of iterations for one run.
#
#
# ------------------------------------------------------------
# tol
# ------------------------------------------------------------
#
# Convergence tolerance.
#
#
# Smaller:
#     stricter convergence
#
#
# ------------------------------------------------------------
# random_state
# ------------------------------------------------------------
#
# Controls randomness.
#
#
# ------------------------------------------------------------
# algorithm
# ------------------------------------------------------------
#
# Algorithm used to compute K-Means.
#
# Common:
#
#     "lloyd"
#     "elkan"
#
#
# ------------------------------------------------------------
# verbose
# ------------------------------------------------------------
#
# Controls output during training.
#
# ============================================================


# ============================================================
# 28. MAX_ITER
# ============================================================

model = KMeans(
    n_clusters=3,
    max_iter=100,
    random_state=42,
    n_init=10
)

model.fit(X)

print(
    "\nNumber of iterations used:",
    model.n_iter_
)


# ============================================================
# 29. TOL
# ============================================================
#
# K-Means stops when centroid movement becomes sufficiently
# small.
#
# tol controls this convergence threshold.
#
# ============================================================

model = KMeans(
    n_clusters=3,
    tol=1e-4,
    random_state=42,
    n_init=10
)

model.fit(X)

print(
    "\nIterations with tol=1e-4:",
    model.n_iter_
)


# ============================================================
# 30. PREDICT NEW DATA
# ============================================================
#
# Once centroids are learned:
#
# New point
#     ↓
# Calculate distance to every centroid
#     ↓
# Choose nearest centroid
#     ↓
# Cluster label
#
# ============================================================

new_data = np.array([
    [-5, -2],
    [1, 4],
    [6, -2],
    [0, 0]
])

new_labels = final_kmeans.predict(
    new_data
)

print(
    "\nNew data:"
)

print(
    new_data
)

print(
    "\nPredicted clusters:"
)

print(
    new_labels
)


# ============================================================
# 31. FEATURE SCALING ⭐⭐⭐
# ============================================================
#
# K-Means is distance-based.
#
# Therefore feature scale matters.
#
#
# Example:
#
# Age:
#
#     20 - 60
#
# Salary:
#
#     20,000 - 200,000
#
#
# Salary can dominate Euclidean distance.
#
#
# Usually scale features when they are on very different
# numerical scales.
#
#
# Common approach:
#
#     StandardScaler
#
# ============================================================


from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

scaled_kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

scaled_labels = scaled_kmeans.fit_predict(
    X_scaled
)

print(
    "\nScaled K-Means inertia:",
    scaled_kmeans.inertia_
)


# ============================================================
# 32. K-MEANS ASSUMPTIONS / LIMITATIONS ⭐⭐⭐
# ============================================================
#
# K-Means works best when clusters are:
#
#     roughly spherical
#     similarly scaled
#     reasonably separated
#
#
# It can struggle with:
#
#     elongated clusters
#     highly overlapping clusters
#     clusters with very different densities
#     many outliers
#
#
# Example:
#
#
#       ● ● ●
#      ● ● ●
#     ● ● ●
#
#
# versus:
#
#
#     ●
#      ●
#       ●
#        ●
#         ●
#
#
# A single centroid may not represent elongated structure
# very well.
#
# ============================================================


# ============================================================
# 33. OUTLIER SENSITIVITY
# ============================================================
#
# K-Means uses squared distances.
#
# Therefore an extreme outlier can strongly affect the
# centroid.
#
#
# Example:
#
#     ● ● ● ●
#        ●
#
#
#                    X
#
# The far-away X can pull the centroid toward itself.
#
# ============================================================


# ============================================================
# 34. ADVANTAGES
# ============================================================
#
# 1. Simple.
#
# 2. Easy to understand.
#
# 3. Fast for many datasets.
#
# 4. Scales reasonably well.
#
# 5. Easy to implement.
#
# 6. Useful for customer segmentation.
#
# 7. Useful for exploratory data analysis.
#
# ============================================================


# ============================================================
# 35. DISADVANTAGES
# ============================================================
#
# 1. Need to choose K.
#
# 2. Sensitive to initialization.
#
# 3. Sensitive to feature scaling.
#
# 4. Sensitive to outliers.
#
# 5. Works poorly for irregular cluster shapes.
#
# 6. Assumes centroid-based cluster structure.
#
# 7. Can converge to a local optimum.
#
# ============================================================


# ============================================================
# 36. K-MEANS vs GMM ⭐⭐⭐
# ============================================================
#
#
# K-MEANS:
#
#     Hard assignment
#
#     Point:
#         Cluster 1
#
#
# GMM:
#
#     Soft assignment
#
#     Point:
#         Cluster 1 = 0.8
#         Cluster 2 = 0.2
#
#
# K-MEANS:
#
#     Uses centroid + distance.
#
#
# GMM:
#
#     Uses probability distributions.
#
#
# K-MEANS:
#
#     Roughly spherical clusters.
#
#
# GMM:
#
#     Can model elliptical clusters using covariance.
#
#
# K-MEANS:
#
#     No probability density.
#
#
# GMM:
#
#     Provides likelihood / probability information.
#
# ============================================================


# ============================================================
# 37. K-MEANS vs HIERARCHICAL CLUSTERING
# ============================================================
#
# K-Means:
#
#     Need K beforehand.
#
#
# Hierarchical:
#
#     Builds a hierarchy/tree of clusters.
#
#     Can inspect different numbers of clusters afterward.
#
#
# K-Means:
#
#     Centroid based.
#
#
# Hierarchical:
#
#     Based on distances and linkage.
#
# ============================================================


# ============================================================
# 38. K-MEANS vs DBSCAN
# ============================================================
#
# K-Means:
#
#     Need K.
#     Centroid-based.
#     Sensitive to outliers.
#
#
# DBSCAN:
#
#     Does not require K.
#     Density-based.
#     Can identify noise/outliers.
#     Can find irregular shapes.
#
# ============================================================


# ============================================================
# 39. INTERVIEW QUESTIONS
# ============================================================
#
#
# Q1. What is K-Means?
#
# Answer:
#
# K-Means is an unsupervised clustering algorithm that
# partitions data into K clusters by repeatedly assigning
# points to their nearest centroid and updating the centroids.
#
#
# ------------------------------------------------------------
#
# Q2. What is a centroid?
#
# Answer:
#
# The centroid is the mean position of the points assigned
# to a cluster.
#
#
# ------------------------------------------------------------
#
# Q3. What are the two main steps of K-Means?
#
# Answer:
#
# Assignment:
#     Assign each point to the nearest centroid.
#
# Update:
#     Recalculate each centroid as the mean of its assigned
#     points.
#
#
# ------------------------------------------------------------
#
# Q4. What is inertia?
#
# Answer:
#
# Inertia is the sum of squared distances between each point
# and its assigned centroid.
#
#
# ------------------------------------------------------------
#
# Q5. Does lower inertia always mean better clustering?
#
# Answer:
#
# Not by itself. Increasing K generally decreases inertia,
# so methods such as the elbow method or silhouette score
# are used to help choose K.
#
#
# ------------------------------------------------------------
#
# Q6. What is K-Means++?
#
# Answer:
#
# It is a smarter centroid initialization method that
# spreads initial centroids apart to improve initialization.
#
#
# ------------------------------------------------------------
#
# Q7. Why use n_init?
#
# Answer:
#
# K-Means can depend on initialization, so multiple runs
# with different initializations can be performed and the
# best solution can be selected.
#
#
# ------------------------------------------------------------
#
# Q8. Why is scaling important?
#
# Answer:
#
# K-Means uses distance, so features with larger numerical
# scales can dominate the distance calculation.
#
#
# ------------------------------------------------------------
#
# Q9. What is the elbow method?
#
# Answer:
#
# It plots inertia against K and looks for a point where
# additional clusters provide diminishing improvement.
#
#
# ------------------------------------------------------------
#
# Q10. What is silhouette score?
#
# Answer:
#
# It measures how similar a point is to its own cluster
# compared with the nearest other cluster. Higher values
# generally indicate better-separated clusters.
#
#
# ------------------------------------------------------------
#
# Q11. K-Means vs GMM?
#
# Answer:
#
# K-Means performs hard centroid-based clustering, while
# GMM performs probabilistic soft clustering using Gaussian
# distributions.
#
#
# ------------------------------------------------------------
#
# Q12. Can K-Means find arbitrary-shaped clusters?
#
# Answer:
#
# Generally no. It works best with roughly spherical,
# centroid-based clusters.
#
#
# ------------------------------------------------------------
#
# Q13. Is K-Means supervised?
#
# Answer:
#
# No. K-Means is unsupervised and does not require class
# labels.
#
#
# ============================================================
# 40. 30-SECOND INTERVIEW EXPLANATION ⭐⭐⭐
# ============================================================
#
# K-Means is an unsupervised clustering algorithm.
# We choose K clusters and initialize K centroids.
# The algorithm repeatedly assigns each point to its nearest
# centroid and then recalculates each centroid as the mean of
# its assigned points.
#
# It minimizes the within-cluster sum of squared distances,
# called inertia. K-Means++ is commonly used for better
# initialization, and the elbow method or silhouette score
# can help choose K.
#
# ============================================================


# ============================================================
# 41. FINAL MEMORY CHEATSHEET
# ============================================================
#
# K-MEANS =
#
# Choose K
#    ↓
# Initialize centroids
#    ↓
# Assign points
#    ↓
# Recalculate centroids
#    ↓
# Repeat
#    ↓
# Converge
#
#
# MUST REMEMBER:
#
# 1. Unsupervised
#
# 2. Clustering algorithm
#
# 3. K = number of clusters
#
# 4. Centroid = mean of cluster points
#
# 5. Assignment = nearest centroid
#
# 6. Update = calculate new mean
#
# 7. Objective = minimize squared distances
#
# 8. Inertia = within-cluster sum of squares
#
# 9. K-Means++ = better initialization
#
# 10. Elbow method = use inertia
#
# 11. Silhouette score = cluster separation/cohesion
#
# 12. Scaling matters
#
# 13. Outliers can affect centroids
#
# 14. Works best for roughly spherical clusters
#
# ============================================================


# ============================================================
# 42. WORKING TREE ⭐⭐⭐
# ============================================================
#
#
#                       K-MEANS
#                          │
#                          ▼
#                       Choose K
#                          │
#                          ▼
#                  Initialize Centroids
#                          │
#                          ▼
#                   Assignment Step
#                          │
#                          ▼
#              Calculate Distance to
#              Every Centroid
#                          │
#                          ▼
#                Choose Nearest
#                  Centroid
#                          │
#                          ▼
#                   Cluster Label
#                          │
#                          ▼
#                     Update Step
#                          │
#                          ▼
#              Calculate Mean of
#              Each Cluster
#                          │
#                          ▼
#                 New Centroids
#                          │
#                          ▼
#                    Converged?
#                     /       \
#                   NO         YES
#                   │           │
#                   └───────────┤
#                               ▼
#                         Final Clusters
#
#
# ============================================================
#
# CHOOSING K:
#
#                    Possible K values
#                          │
#                 ┌────────┴────────┐
#                 │                 │
#                 ▼                 ▼
#             Elbow Method     Silhouette Score
#                 │                 │
#                 ▼                 ▼
#              Inertia          Separation
#                 │                 │
#                 └────────┬────────┘
#                          ▼
#                       Choose K
#
#
# ============================================================
#
# ONE-LINE MENTAL MODEL:
#
#     "Keep moving each point to its nearest center,
#      then move each center to the mean of its points,
#      until nothing changes."
#
# ============================================================
```
